# 1. Key Terms

| Term | Meaning |
|---|---|
| **Terminal** | A program for interacting with a computer's files and executing programs through a command line interface (CLI). |
| **Shell** | A program running inside the terminal that reads what you type, interprets it as a command, and executes it. |
| **Command Line Interface (CLI)** | A type of user interface (UI) that lets a user perform actions by entering text-based commands. |
| **Graphical User Interface (GUI)** | A type of user interface that uses visual elements such as icons, buttons, windows, and dialog boxes, allowing actions like clicking and drag-and-drop. |
| **Directory** | Another term for a "folder": it contains references to files and possibly other directories. |
| **Working Directory** | The directory where your commands will be executed. |
| **Command** | A single action to be performed on your computer, such as creating a file, listing a directory's contents, navigating to another directory, or executing a program. |
| **Argument** | An additional piece of information provided to a command to change the command's behavior. |
| **Python** | The programming language we will be writing in and learning. |
| **Interpreter** | The program that reads a Python file and executes it, one statement at a time. |
| **python3** | The command that runs the Python interpreter on a file. |

## How terminal, shell, and CLI fit together

**My question:** Is it okay to say the terminal is the program/window that provides the CLI, the shell interprets the text and talks to the operating system, and the text-based interface we are using is the CLI?

**Answer:** Yes, that is correct.

- The **terminal** is the window you type into.
- The **shell** (bash or zsh) is the program inside it that reads each line, works out what it means, and asks the operating system to do it.
- The **CLI** is the style of interface: text commands in, text output back.

```
You type a command  ->  Terminal (window)  ->  Shell (interprets)  ->  Operating system (does the work)
```

# 2. `pwd` and `ls`

In the terminal you interact with one directory at a time: the **working directory**. It is the directory you are currently in and running commands against.

## pwd (print working directory)
Prints the full file path to the working directory.

```bash
pwd
```

## ls (list)
Prints the contents of the working directory.

```bash
ls
```

`-l` shows detailed information for each file, such as size and permissions.

```bash
ls -l
```

## ls ..
Prints the contents of the **parent directory**, but does **not** move you there.

The parent directory is the folder one level above the working directory.

```bash
ls ..
```

# 3. Navigating with `cd`

> **Tip:** Use the `Tab` key to autocomplete commands and file names. Start typing and press `Tab`.

## cd [directory]
Moves you to another directory in the file system. The destination is the **argument** for `cd`.

```bash
cd Documents
```

## Going more than one level at once
Use a slash `/` between directory names. For example, from `/Users` you can reach `/Users/smith/Documents` in one command:

```bash
cd smith/Documents
```

## The parent directory: `..`
The special name `..` always refers to the parent of the working directory.

```bash
cd ..        # move up one level
ls ..        # list the parent's contents without moving
cd ../..     # move up two levels
```

## cd with no argument
Takes you to your **home directory** (shown as `~`), for example `/Users/smith`.

```bash
cd
```

> **Correction to my original notes:** `cd` alone goes to the *home* directory, not the *root*. The root is the very top of the file system, written `/`, and you get there with `cd /`.

# 4. Making Files and Directories: `mkdir` and `touch`

## mkdir [dir_name]
Creates a new directory inside the working directory.

```bash
mkdir projects
```

## touch [file_name]
Creates a new, empty file in the working directory. Include the file extension (like `.py` or `.txt`) so the computer knows what kind of file it is.

```bash
touch notes.txt
```

## Creating several at once
List multiple names with a space between them.

```bash
touch file1.txt file2.txt file3.txt    # creates 3 files
mkdir dir1 dir2 dir3                   # creates 3 directories
```

# 5. Running Python Files: `python3`

A Python program is any file with a `.py` extension, like `script.py`.

## python3 [file_name.py]
Runs the file using the Python interpreter.

```bash
python3 script.py
```

## Control + C

# 6. `echo`, `>>`, `>`, and Unfinished Quotes

## echo
Prints the given string straight to the terminal.

```bash
echo "hello world"
```

`-n` makes `echo` not print a new line at the end.

```bash
echo -n "hello world"
```

## >> (append)
Sends the output to a file instead of the terminal, adding it to the **end** of the file. If the file doesn't exist, it is created.

```bash
echo "hello world" >> output.txt
```

## > (overwrite)
Empties the file first, then writes. If the file doesn't exist, it is created.

```bash
echo "hello world" > output.txt
```

| Operator | If the file already has content |
|---|---|
| `>>` | Keeps it and adds the new text at the end |
| `>` | **Erases it** and writes only the new text |

## Unfinished quotes
If you open a quote and never close it, the shell waits for the rest. The prompt changes to a continuation marker, and everything you type is treated as more of the same string.

```bash
echo "hello world
>
```

This also happens when the quotes don't match, like starting with `'` and ending with `"`.

**Fix:** type the missing closing quote and press Enter, or press `Control + C` to cancel and start over.

# 7. `cat` and Combining Commands with `&&`

## cat [file_name]
Reads a file and prints its contents to the terminal.

```bash
cat output.txt
```

You can give `cat` several files and it prints them one after another.

```bash
cat file1.txt file2.txt
```

Mixed with `>`, you can combine files into a new one.

```bash
cat file1.txt file2.txt > combined.txt
```

## && (and)
Combines two commands on one line. The second runs **only if the first succeeds**; if a command fails, the ones after it do not run.

```bash
ls && cat output.txt
```

This lists the contents of the working directory, then prints the contents of `output.txt`.

# 8. Removing, Renaming, Moving, and Copying

## rm [file_name]
Removes a file from the working directory.

```bash
rm notes.txt
```

## rm -r [directory_name]
Removes a directory and all of its contents. `-r` stands for "recursive".

```bash
rm -r projects
```

> **Warning:** `rm` deletes permanently. Nothing goes to the Trash, and there is no undo.

## mv (move or rename)
Renames a file, or moves it to another directory.

```bash
mv old_name.txt new_name.txt    # rename
mv notes.txt projects           # move into the projects directory
```

## cp [file_name] [destination]
Copies a file. The original stays where it is.

```bash
cp notes.txt projects           # copy into the projects directory
cp notes.txt notes_backup.txt   # make a copy with a new name
```

> **Correction to my original notes:** `cp` **copies** a file; it does not move it. `mv` is the command that moves.

# 9. Quick Reference

| Command | What it does |
|---|---|
| `pwd` | Print the working directory |
| `ls` | List the contents of the working directory |
| `ls -l` | List with details (size, permissions) |
| `ls ..` | List the parent directory's contents |
| `cd [directory]` | Move into a directory |
| `cd ..` | Move up to the parent directory |
| `cd ../..` | Move up two levels |
| `cd` | Go to the home directory |
| `mkdir [name]` | Make a new directory |
| `touch [name]` | Make a new file |
| `python3 [file.py]` | Run a Python file |
| `Control + C` | Stop the running program or cancel the current line |
| `echo "text"` | Print text to the terminal |
| `echo "text" >> file` | Append text to a file |
| `echo "text" > file` | Overwrite a file with text |
| `cat [file]` | Print a file's contents |
| `cmd1 && cmd2` | Run `cmd2` only if `cmd1` succeeds |
| `rm [file]` | Delete a file (permanent) |
| `rm -r [directory]` | Delete a directory and everything in it (permanent) |
| `mv [file] [new_name]` | Rename a file |
| `mv [file] [directory]` | Move a file |
| `cp [file] [destination]` | Copy a file |
| `Tab` | Autocomplete a command or file name |
