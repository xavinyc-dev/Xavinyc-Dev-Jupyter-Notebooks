# 1. Key Terms

| **Terms** | **Definitions** |
|---|---|
| **Loops** | A way to make Python repeat code multiple times, making it easy to complete repetitive tasks. |
| **`for` loop with `range()`** is used to do repetitive tasks a known number of times like counting to 10. |
| **Iterable** | An iterable is any value a for loop can go through, one element at a time. A `range()` gives a number one a time and a string gives one character at a time. They are both iterables, anything after in is usually an iterable. |
| **`while loop`** | This is a loop that repeats a process if the condition is True an unknown number of times but continuing until the condition is False. |
| **Infinite Loops** | These occur when the loop's condition never becomes False; you use `break` to exit early or `continue` to skip to the next iteration. |
| **Nested Loops** | are loops that are inside of loops, these are useful for grids, rows, and columns. |
| **Loop challenges** | Loop challenges help you practice using loops to solve real problems, such as counting results or repeating until something happens. |

# 1. Intro to Iterations

Iteration means doing the same steps over and over, where each round gets closer and closer to the finish. Each single round is called an iteration, one pass through a loop. Every pass changes something.

Picture this: you are assigned the task of rolling a 6-sided die 100 times and documenting the result of each.

```py
1. a 4 was rolled
2. a 2 was rolled
3. a 6 was rolled
4. a 6 was rolled
5. a 1 was rolled
...

1. Gather your materials (pen, paper, a die)

2. Until you've rolled 100 times you will:
    
    a. roll the die

    b. record the result by writing: "[roll number]. a [roll result] was rolled"

3. Make sure to increase the roll number after each roll.

````python
import random

print(f"1. a {random.randint(1, 6)} was rolled")
print(f"2. a {random.randint(1, 6)} was rolled")
print(f"3. a {random.randint(1, 6)} was rolled")
print(f"4. a {random.randint(1, 6)} was rolled")
print(f"5. a {random.randint(1, 6)} was rolled")
# and so on until you reach 100! 🫠
````

The first line, `import random`, loads a tool that comes with Python, and `random.randint(0,6) returns a random integer between 1 and 6 (including 1 and 6). # 3. The `for` loop

`for` loops are for repeating code for a known number of times. Paired with the `range()` function the syntax looks like this:

```py
for i in range(how_many_times):
    # do something
```

Using this syntax we can now easily loop the statement before 100 times

````python
for i in range(100):
    print(f"{i}. a {random.randint(1, 6)} was rolled")
````

`range(100)` produces the numbers 0 to 99 and `i` is the variable assigned each number one at a time, one number per loop. 100 numbers, 100 values for `i`.

1. Take the next number from the range and assign it to i

2. Execute the code block

3. If the range has more numbers, go to step 1

4. Otherwise, the loop ends and the program continues with the first line after the loop.
