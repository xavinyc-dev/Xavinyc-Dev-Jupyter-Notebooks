# 1. Key Terms

| Term | Meaning |
|---|---|
| **Repository (repo)** | A centralized location where files are stored and managed. Any folder can be considered a repository. |
| **Git** | A "version control system" that lets us manage the history of changes made to a repo through commits. |
| **Commit** | A "snapshot" of the changes made to a repo. Typically created when a key milestone is reached, for example when a feature is completed. |
| **Snapshot** | A record that saves the complete state of all files at that specific time. |
| **Staging Area** | A place to temporarily store changed files to include in the next commit. |
| **GitHub** | An online host of Git repositories, with tools for managing Git projects and features for collaboration. |
| **Local Repository** | A repository stored on a developer's computer. |
| **Remote Repository** | A repository stored online on a service like GitHub. |
| **Clone** | Copy a remote repo's files and commit history and store them locally (creates a local repository). |
| **Push** | Send a local repo's commit history to a remote repo so the two are synchronized. |

# 2. What is Git?

Git is a tool that takes snapshots of the files in a project, which lets developers manage them and, if needed, revert to prior versions. It is an example of a **version control system**: it helps manage the versions of our projects.

- Git maintains a history of changes through **commits**.
- A commit is a snapshot of our code at that exact moment. It represents a completed unit of work, such as a new feature or a bug fix.

The commits are stored alongside the code files. Together, the files and the commit history form a **Git repository**.

## Repositories
A repository (or "repo") is a centralized location where data is stored and managed. Any folder with files can be considered a repo.

A **Git repository** is simply a folder whose files are managed by Git.

---
# 3. What is GitHub?

GitHub is a cloud-based "host" for Git repositories. In other words, it stores Git repositories online. GitHub lets developers store, share, and collaborate on their projects.

| | Git | GitHub |
|---|---|---|
| **What it is** | A tool on your computer | A website |
| **What it does** | Tracks the history of changes | Stores repos online and helps people collaborate |
| **Where the repo lives** | Local repository | Remote repository |

# 4. The Git Workflow

Using Git involves moving our code between three phases:

1. **The working directory**: where we are editing the files
2. **The staging area**: where we temporarily store changes before committing them
3. **The repository**: where we store our committed changes

```
Working directory  --git add-->  Staging area  --git commit-->  Repository
 (editing files)                 (ready to save)               (saved in history)
```

# 5. `git init` and `git status`

## git init
To use Git in a project, you must turn the project into a Git repository. Run this inside the project folder:

```bash
git init
```

You only do this once per project. If you cloned the repo from GitHub, it is already a Git repository and you skip this step.

## git status
Shows which files have changed, which are staged, and which are new.

```bash
git status
```

| What `git status` shows | Meaning |
|---|---|
| **Untracked files** (red) | New files Git is not tracking yet |
| **Changes not staged for commit** (red) | Edited files that have not been added yet |
| **Changes to be committed** (green) | Staged files that will go into the next commit |

# 6. `git add`, `git commit`, and `git log`

Once we have changes we want to save in the repository:

## git add
"Stages" the changes, moving them into the staging area.

```bash
git add notes.txt    # stage one file
git add -A           # stage all changed files in the whole repo
git add .            # stage all changed files in the current folder and below
```

> **Correction to my original notes:** the flag is `-A` (dash first, capital A), not `a-`. Lowercase `-a` does not work with `git add`.

## git commit
Saves the staged changes as a new commit. `-m` adds a message describing what changed.

```bash
git commit -m "Add string methods notes"
```

Use the same kind of quote at the start and end of the message. If they don't match, the terminal waits for more input; press `Control + C` to cancel.

## git log
Shows the commit history and details: who made each commit, when, and the message.

```bash
git log
```

Press `q` to exit the log. For a shorter view with one line per commit:

```bash
git log --oneline
```

# 7. The GitHub Workflow

1. **Create a new repo on GitHub.** This is a remote repository.
2. **Clone (copy) the repo from GitHub onto our computer.** Now we have a local repository that is linked to the remote repository.
3. **Use the normal Git workflow:** make changes > `git add` > `git commit`
4. **Push the changes.** After making commits on our local repository, we push them to the remote repository.

## git clone
Downloads a repository from GitHub. It needs the repo's URL, which you copy from the green **Code** button on the repo's GitHub page.

```bash
git clone [repository URL]
```

## git push
Uploads local commits to the remote repository on GitHub.

```bash
git push
```

## git pull
Downloads new commits from the remote repository into your local one. Use it when the GitHub copy has changes your computer doesn't have yet.

```bash
git pull
```

## The whole cycle

```bash
git status                        # see what changed
git add -A                        # stage the changes
git commit -m "Describe the work" # save a snapshot
git push                          # upload to GitHub
```

```
GitHub (remote)  --git clone / git pull-->  Your computer (local)
GitHub (remote)  <--------git push--------  Your computer (local)
```

# 8. Summary

**Git** is a "version control system" that lets us manage the history of changes made to a repo. Think of the Camera app on your phone, documenting the world around you.

**GitHub** is an online host of Git repositories, with tools for managing Git projects and features for collaboration. Think of where your photos are stored online.

Together, we can manage code in local repositories and sync it with remote repositories.

# 9. Quick Reference

| Command | What it does |
|---|---|
| `git init` | Add Git to the current folder (make it a repo) |
| `git status` | Check the status of changed files in the repo |
| `git add [filename]` | Add the given file to the staging area |
| `git add -A` | Add all changed files to the staging area |
| `git commit -m "[message]"` | Create a new commit from the staged files |
| `git log` | Show the commit history |
| `git push` | Upload local commits to the remote (GitHub) |
| `git pull` | Download new commits from the remote (GitHub) |
| `git clone [URL]` | Download a Git repository from the remote (GitHub) |
