### Terms used in this notebook
| Term | Meaning |
|---|---|
| **string** | Text data, written inside quotes: `'hello'` or `"hello"`. |
| **variable** | A name that stores a value: `email = 'a@b.com'`. |
| **method** | A function that belongs to a value, called with a dot: `text.upper()`. |
| **index** | The position of a character in a string. Python counts from **0**. |
| **substring** | A smaller piece of a string, e.g. `'@'` or `'.com'` inside an email. |
| **boolean** | A value that is either `True` or `False`. |
| **list** | An ordered collection of values in square brackets: `['a', 'b', 'c']`. |
| **f-string** | A string with `f` in front so you can put variables inside `{}`: `f'Hi {name}'`. |
| **chaining** | Calling methods one after another: `text.strip().lower()` — runs left to right. |
| **immutable** | Strings can't be changed in place. Methods like `.upper()` return a **new** string; the original stays the same. |

## 1. Strings and f-strings (Steps 1–2)

An **f-string** lets you drop a variable straight into text. Put an `f` before the opening quote and
wrap the variable in curly braces `{}`.

```python
email = 'alice.johnson@company.com'
print(f'Email: {email}')
```

## 2. Finding a character with `.find()` (Step 3)

`.find(substring)` returns the **index** (position) of the **first** place the substring appears.

- Counting starts at **0**, so in `'hello'` the `'e'` is at index 1.
- If the substring isn't there, `.find()` returns **`-1`** (it does not crash).

```python
at_position = email.find('@')
print(f'Position of @: {at_position}')

# Extra: what happens when it's not found
print('hello'.find('z'))  # -1
```

## 3. Slicing to pull out part of a string (Step 4)

**Slicing** grabs a range of characters: `text[start:stop]`.

- `start` is included, `stop` is **not** included.
- Leaving `start` empty means "from the beginning": `text[:5]` = first 5 characters.
- Leaving `stop` empty means "to the end": `text[6:]`.

Here, `email[:at_position]` takes everything **before** the `@`, which is the username.

```python
username = email[:at_position]
print(f'Username: {username}')

# Extra: everything AFTER the @ (the domain) — +1 skips the @ itself
print(email[at_position + 1:])  # company.com
```

## 4. Checking if something exists with `in` (Steps 5–6)

`in` is a **keyword** (not a method — no dot, no parentheses). `substring in text` gives a **boolean**:
`True` if it's there, `False` if not. Good for quick validation checks.

```python
has_at = '@' in email
print(f'Has @ symbol: {has_at}')

has_com = '.com' in email
print(f'Has .com: {has_com}')
```

## 5. Cleaning messy text: `.strip()` and `.title()` (Steps 7–9)

Real data often has extra spaces and random capitals.

- **`.strip()`** removes whitespace (spaces, tabs, newlines) from **both ends** — not from the middle.
- **`.title()`** capitalizes the first letter of every word and lowercases the rest.

Chaining `messy_name.strip().title()` means: strip first, then title-case the result.

*Note:* the single quotes inside the f-string are just there so you can **see** the spaces. Since the
f-string itself uses double quotes `"..."`, single quotes inside it are fine.

```python
messy_name = '  sARaH dAVis  '
print(f"Original name: '{messy_name}'")

clean_name = messy_name.strip().title()
print(f"Cleaned name: '{clean_name}'")
```

## 6. Normalizing case: `.lower()` and `.upper()` (Steps 10–11)

- **`.lower()`** → all lowercase. Used to **normalize** emails so `SARAH@X.COM` and `sarah@x.com` count as the same.
- **`.upper()`** → all uppercase. Handy for headings or name badges.

Reminder: strings are **immutable** — `messy_email` is unchanged; `clean_email` is a new string.

```python
messy_email = '  SARAH.DAVIS@COMPANY.COM  '
clean_email = messy_email.strip().lower()
print(f"Original email: '{messy_email}'")
print(f"Cleaned email: '{clean_email}'")

name_badge = clean_name.upper()
print(f'Name badge: {name_badge}')
```

## 7. Reusing `.find()` + slicing on the cleaned email (Step 12)

Same technique as Sections 2–3, now on `clean_email`.

```python
at_pos = clean_email.find('@')
email_user = clean_email[:at_pos]

print(f'Email: {clean_email}')
print(f'Username: {email_user}')
```

## 8. Swapping or removing characters with `.replace()` (Steps 13–14)

`.replace(old, new)` replaces **every** occurrence of `old` with `new`.

- Replace with a space: `'sarah.davis'.replace('.', ' ')` → `'sarah davis'`
- **Remove** a character by replacing it with an empty string `''`.

```python
display_name = email_user.replace('.', ' ').title()
print(f'Display name: {display_name}')

phone = '123-456-7890'
phone_clean = phone.replace('-', '')
print(f'Phone with dashes: {phone}')
print(f'Phone without dashes: {phone_clean}')
```

## 9. Breaking apart and putting back together: `.split()` and `.join()` (Steps 15–16)

- **`.split(separator)`** cuts a string into a **list** wherever the separator appears.
  The separator itself is removed. Here it's `', '` (comma **and** space), so no leftover spaces.
- **`.join(list)`** is the opposite: glues a list of strings into one string.
  ⚠️ The syntax looks backwards — the **separator** comes first: `' | '.join(parts)`.

```python
full_address = '123 Main Street, Springfield, IL'
address_parts = full_address.split(', ')
print(f'Address parts: {address_parts}')

rejoined = ' | '.join(address_parts)
print(f'Rejoined: {rejoined}')

# Extra: lists use indexes too
print(address_parts[1])  # Springfield
```

## 10. Checking the start and end: `.startswith()` and `.endswith()` (Steps 17–19)

Both return a **boolean**.

- `.startswith(prefix)` → does the string **begin** with this?
- `.endswith(suffix)` → does the string **end** with this?

More precise than `in`: `'.com' in email` is `True` even for `'x@site.com.fake'`,
but `.endswith('.com')` would catch that it doesn't actually end in `.com`.

```python
employee_code = 'EMP-2024-SD'
starts_with_emp = employee_code.startswith('EMP')
ends_with_com = clean_email.endswith('.com')

print(f'Code: {employee_code}')
print(f'Starts with EMP: {starts_with_emp}')
print(f'Clean email: {clean_email}')
print(f'Ends with .com: {ends_with_com}')
```

## 11. Counting with `.count()` (Steps 20–21)

`.count(substring)` returns how many times the substring appears (non-overlapping).
`'sarah.davis@company.com'` has 2 dots; `'EMP-2024-SD'` has 2 dashes.

```python
dot_count = clean_email.count('.')
dash_count = employee_code.count('-')

print(f'Dots in email: {dot_count}')
print(f'Dashes in code: {dash_count}')
```

## Quick reference

| Method / tool | What it does | Returns | Example → result |
|---|---|---|---|
| `f'{var}'` | Put a variable inside text | string | `f'Hi {name}'` → `'Hi Sarah'` |
| `.find(x)` | Index of first `x` | int (`-1` if missing) | `'hello'.find('e')` → `1` |
| `text[a:b]` | Slice from `a` up to (not incl.) `b` | string | `'hello'[:2]` → `'he'` |
| `x in text` | Is `x` inside? | bool | `'@' in 'a@b'` → `True` |
| `.strip()` | Remove spaces at both ends | string | `'  hi  '.strip()` → `'hi'` |
| `.title()` | Capitalize Each Word | string | `'sARaH'.title()` → `'Sarah'` |
| `.lower()` / `.upper()` | All lower / all upper | string | `'Hi'.upper()` → `'HI'` |
| `.replace(a, b)` | Swap every `a` for `b` | string | `'1-2'.replace('-', '')` → `'12'` |
| `.split(sep)` | String → list | list | `'a, b'.split(', ')` → `['a', 'b']` |
| `sep.join(list)` | List → string | string | `'-'.join(['a','b'])` → `'a-b'` |
| `.startswith(x)` / `.endswith(x)` | Begins / ends with `x`? | bool | `'EMP-1'.startswith('EMP')` → `True` |
| `.count(x)` | How many `x` | int | `'banana'.count('a')` → `3` |

**Key idea:** none of these change the original string — they all hand back a new value, so store it in a variable.
