# 1. Key Terms

| **Term** | **Definition** |
|---|---|
| **Lists** | Ordered collections of values separated by commas, called **elements**. Have indexes like `strings` starting at `0`, support bracket notation `[]` and `slicing`, and work with `len()` and the `in` operator.<br>• Lists are **mutable** so methods like `append`, `insert`,`pop`, and `remove` change a list in place instead of creating another.<br>• Assigning another variable to it just adds another label. |
| **Parallel lists** | Two or more lists where the same index describes the same thing, like `items[0]` and `prices[0]` both describing the first purchase. |
| **Mutate** | To change an object's contents in place. The variable still references the same object. |
| **Reference** | A variable does not hold a value, it's a name that points to an object in memory, rather than being the object itself. It holds a **reference** to the value. Two variables can reference the same list, and a function receiving a list receives a reference to the caller's list. |
| **Alias** | Two variables that reference the same list are **aliases**, a change made through one of them shows up through the other. |
| **Copying a list** | With `list(items)` or `items[:]`  makes a new list with the same elements. A function that copies a list before changing it leaves the caller's list unchanged. |
| **Unpacking** | Assigns the elements of a list to several variables in one statement. e.g. `item, price = ["coffee", 3.50]`. |
| **`enumerate()`** | A function that produces `(index, item)` pairs that can be iterated over with a `for` loop. It usually unpacks each pair into two variables, separated by commas. `for index, item in enumerate(items):`. |
| **2D lists** | Lists that contain other lists. The first index picks the row and the second picks the column: `receipt[1][0]`. |
| **Tuple** | Like a list that cannot change. It's immutable unlike lists. They are written with parentheses `()`. |

### Tutor note: a few definitions to keep precise

- **One list, many elements:** a list is one Python object containing an ordered collection of elements. A variable refers to the whole list; an index selects an element. Even an empty list is still a list.
- **Ordered** means elements have positions. It does not mean Python automatically sorts them.
- **Mutable** means the existing list's contents can change. It does not mean every operation changes it: slicing reads a list and produces a new outer list.
- **Alias:** two names for the same object. A mutation can be observed through both names; reassigning one name does not reassign the other.
- **Tuple:** a separate sequence type whose element references cannot be replaced, added, or removed after creation. It is similar to a list, but is not literally a list. See Section 6.
- **Copying:** `list(items)` and `items[:]` create a new outer list. These are shallow copies: nested mutable objects may still be shared. There's a tutor note on this in the 2D Lists section.

You do not need to master every term in the table before continuing. Learn it again when you meet a concrete example.

# 2. Keeping Every Value



```python
total = 0

while True:
    amount = input("Enter an amount in dollars (like 4.50), or done to finish: ").strip()

    # exit on "done"
    if amount.lower() == "done":
        break

    # skip a blank answer
    if amount == "":
        continue

    total += float(amount)

print(f"Your total is ${total:.2f}")
```

Above is a dollar program that adds dollar amounts entered by a user to a `total` and prints it out.

We want a full receipt which lists every item the customer bought and what each one costs.

A **list** is a single value that holds any number of values in order and a program can add to it one value at a time. The program below keeps two lists, one for the names of the items and one for their prices, and prints a receipt when the user types `done`.

````python
items = []
prices = []

while True:
    # now we also ask for an item (or done)
    item = input("Item (or done to finish): ").strip()
    if item.lower() == "done":
        break

    # then we ask for the price (for now, assume the user enters a valid number)
    price = float(input(f"Price of {item}: "))

    # add the item to the items list and price to the prices list
    items.append(item)
    prices.append(price)

# Once the while loop is done collecting inputs, calculate the total and print out the receipt
total = 0
for i in range(len(items)):
    print(f"{items[i]}: ${prices[i]:.2f}")
    total += prices[i]

print(f"Total: ${total:.2f}")
````

# 3. List Basics

A list is written with square brackets `[]` around its values, with the values separated by commas `,`.
```python
items = ['coffee', 'bagel', 'juice', 'muffin']
prices = [3.50, 2.25, 4.00, 3.00]
```
Each value before the comma is called an **element of the list**. Each element stays in its position until the list is changed.


Like strings, lists also:

- have indexes starting at `0`
- use bracket notation `[]` to access individual elements: `my_list[index]`
- work with `len()` to get the number of elements in the list
- support slicing with `[start:end]` and the `in` operator
- unlike strings, lists are mutable

````python
items = ['coffee', 'bagel', 'juice', 'muffin']
print(f"type() on a list returns {type(items)}") # returns the class which is list

print(f"I bought {len(items)} items") # returns amount of elements: 4
print(f"My first item was {items[0]}") # returns the first element in index 0 coffee
print(f"My last item was {items[-1]}") # returns the last element in index -1 muffin
print(f"My first two items were {items[0:2]}") # slices and returns elements from index 0 to and not including 2 coffee bagel
print(f"Did I buy a bagel? {'bagel' in items}") # searches for 'bagel' as an element in items
print('Here is everything I bought:')

for item in items:
    # item holds the "current" item that we're looking at in this iteration of the loop
    print(item)
````

`for item in items` visits each element in order and assigns it to `item`, with no index and no `range()` in sight. The receipt loop needed `range(len(items))` because it also had to get the matching price from a second list.

### Tutor note: length, positions, and choosing a loop

For four elements, the length is 4 and the nonnegative indexes are **0, 1, 2, 3**. Index 4 is outside the list. Index -1 selects the last element only when the list is nonempty.

| Operation | What it gives you | Changes the original list? |
|---|---|---|
| `items[0]` | One element | No |
| `items[0:2]` | A new list containing positions 0 and 1 | No |
| `len(items)` | Number of elements | No |
| `'bagel' in items` | A Boolean membership result | No |
| `for item in items` | Each element, one at a time | The loop itself does not mutate it |
| `for i in range(len(items))` | Each nonnegative index | The loop itself does not mutate it |

Use a direct loop when you need each element. Use indexes when positions matter, such as matching a second list. Later, `enumerate()` provides both the index and the element.

**A useful difference from strings:** `'bag' in 'bagel'` is `True` because it searches for a substring, but `'bag' in ['bagel']` is `False` because a list only matches an equal whole element.

**Empty-list behavior:** `len([])` is 0; a direct loop over it has no passes; accessing its first element raises `IndexError`. A slice can safely produce an empty list instead.

**Challenge**: This function can verify whether or not a given list items contains a given value value. For each numbered comment below, add a comment explaining the logic of the function!

````python
def has_value(items, value):
    for item in items:  # 1. Check every element which is for element in list items, Each passthrough is the next element
        if item == value:  # 2. If that element in the check is the same as the value given
            return True  # 3. return true
    return False  # 4. Once its finished looping it returns False since it didn't find anything to return True

items = ['coffee', 'bagel', 'juice', 'muffin']
print(has_value(items, 'bagel'))   # Prints True
print(has_value(items, 'cookie'))  # Prints False
````

### Tutor note: what `return` does inside a loop

`return` doesn't just give back a value. It **ends the function immediately**, even in the middle of a loop. So `has_value(items, 'coffee')` returns `True` on the very first pass and never checks the rest.

That's also why `return False` has to sit **after** the loop, not inside it. If it were indented inside the `for`, the function would return `False` as soon as the first element didn't match, without checking the others. `has_value(items, 'bagel')` would wrongly return `False` because `'coffee'` is checked first.

| Placement | Meaning |
|---|---|
| `return True` inside the `if` | "Found it, stop looking" |
| `return False` after the loop | "Checked everything, it isn't here" |

**Shortcut:** Python's `in` operator does exactly this job, so `'bagel' in items` gives the same result as `has_value(items, 'bagel')`. Writing it yourself shows what `in` is doing behind the scenes: checking each element in order until one matches.

## Lists Are Mutable

Whenever slicing or using a method on a string, they both return a new string and leave the original alone. If we do not store the new string in a variable the new string is gone. 

````python
item = 'bagel'
item.upper()   # doesn't mutate the string, it makes a copy
item[0] = 'B'  # TypeError: 'str' object does not support item assignment
````

We can change the contents of a list without reassigning them which is called **mutating**. If bagel was on sale, the program can fix its price in place.

````python
prices = [3.50, 2.25, 4.00, 3.00]
prices[1] = 1.75
print(prices)  # [3.5, 1.75, 4.0, 3.0]
````

We can even empty a list in one go using `list.clear()`.

````python
prices.clear()  # prices now has no values
print(prices)  # []
````

### Tutor note: mutation and reassignment answer different questions

**Mutation:** what changed inside the existing object?

**Reassignment:** which object does this variable name refer to now?

In `prices[1] = 1.75`, the square brackets select a slot inside the existing list. The list keeps its identity, while that slot refers to a different number. Python did not turn the number 2.25 into 1.75; it replaced the reference in the slot.

In `prices.clear()`, the same list becomes empty. In `prices = []`, the name is assigned a new empty list. They can look alike when printed, but aliases make the difference observable. You will explore that in References.

**About the intentional string error above:** `item[0] = 'B'` raises `TypeError` because strings do not allow item assignment. A notebook command that runs all cells may stop there. This is an expected demonstration, not evidence that you failed. Decide yourself when to run that demonstration separately.

**`item.upper()` without assignment:** printing `item` afterwards still shows `'bagel'`. `upper()` built a new string `'BAGEL'`, and since nothing stored it, it was thrown away. To keep it: `item = item.upper()`.

## Adding and Removing Values

Since lists are mutable they also have methods for adding to and removing from the list.

Methods to add values to list:

- `my_list.append(value)` -- adds a given value to the end of a list, increasing the length by 1.

- `my_list.insert(index, value)` -- adds a given value at the given index, pushing everything after it one place to the right.

Methods to remove values from list:

- `my_list.pop()` -- removes and returns the last element of a list, reducing the length by one.

- `my_list.pop(index)` -- removes and returns the element at the given index

- `my_list.remove(value)` -- removes the first element equal to the given value

- `my_list.clear()` -- removes every element

````python
items = ['coffee', 'bagel', 'juice', 'muffin']

items.append('cookie')      # adds 'cookie' to the end
items.insert(0, 'tea')      # adds 'tea' to the beginning
print(items)  # Prints ['tea', 'coffee', 'bagel', 'juice', 'muffin', 'cookie']

items.pop()                 # removes the last item ('cookie')
items.pop(0)                # removes the first item ('tea')
print(items)  # Prints ['coffee', 'bagel', 'juice', 'muffin']

items.insert(2, 'scone')    # At index 2, inserts 'scone'
print(items)  # Prints ['coffee', 'bagel', 'scone', 'juice', 'muffin']

items[2] = 'croissant'      # At index 2, replaces 'scone' with 'croissant'
print(items)  # Prints ['coffee', 'bagel', 'croissant', 'juice', 'muffin']

items.remove('croissant')   # Removes the first 'croissant' it finds
print(items)  # Prints ['coffee', 'bagel', 'juice', 'muffin']
````

Each of the methods above changes one list rather than making a copy.

### Tutor note: changing a list versus returning a value

A method can change an object, return a value, do both, or do neither. Ask these questions separately.

| Tool | How it chooses a position | What it returns |
|---|---|---|
| `.append(value)` | Adds at the end | `None` |
| `.insert(index, value)` | Adds before the specified position | `None` |
| `.pop()` | Removes the last element | The removed element |
| `.pop(index)` | Removes the specified position | The removed element |
| `.remove(value)` | Removes the first equal element | `None` |
| `.clear()` | Removes every element | `None` |

All of these mutate the list. Inserting shifts later elements right; removing shifts later elements left. Replacing a slot with bracket assignment does not change the length.

**Common misunderstanding:** assigning the result of `.append()` back to the list name would assign `None` to that name. The append already changed the existing list; it does not return that list.

**Errors to recognize:** popping an empty list or an invalid position raises `IndexError`. Removing a value that is absent raises `ValueError`.

**Example:** with `playlist = ['intro', 'song', 'song']`, `playlist.remove('song')` removes only the **first** `'song'` (index 1). `playlist.pop()` instead returns the last `'song'`. Either way, `['intro', 'song']` remains, but `remove` searched by value while `pop` went by position.

**Parallel-list warning:** if `items.remove('bagel')` runs without removing the matching price, every item after it shifts left but the prices don't. `'juice'` ends up next to the bagel's price, and the receipt is wrong from that point on. The fix is to always change both lists together at the same index. Dictionaries (1.8) solve this by keeping each name and price together.

# 4. References

Lists and dictionaries are considered **mutable types**, which changes how variables that hold them behave.

## A List is a Row of References

Python stores every value in an area of memory called the **heap**. A variable does not hold the value itself. It's like a label that **references the value's heap address**, a note saying where in memory the value is stored.

A list works the same way, one level deeper. The list itself is a row of numbered slots, and each slot holds a reference to one element:

```text
items ──► [ slot 0 | slot 1 | slot 2 ]
               │        │        │
               ▼        ▼        ▼
           'coffee'  'bagel'  'juice'
```

The variable `items` references the list, and each slot references an element. Bracket notation follows those references: `items[1]` goes to the list that `items` references, then follows slot `1` to `'bagel'`.

**For my notes:** A list contains an ordered sequence of references to its elements.

**Predict before running:** What value would `items[2]` give you in this diagram?


### Tutor note: draw the arrows before thinking about memory addresses

An **object** is a Python value, such as a list or a string. A **reference** is a connection to that object. A variable name is bound to an object; a list contains references to its elements.

Imagine an unrelated reading list: one name, `books`, refers to a list containing `'Dune'` and `'Coraline'`. Assigning `reading = books` gives the same list another name.

```text
books   ──┐
          ├──► one list: ['Dune', 'Coraline']
reading ──┘
```

There are two names but only one list. Assignment alone did not make a second list.

- A mutation through either name is visible through both, because both reach the same list.
- Reassigning `reading` to a new list moves only that name's connection. It does not move `books`.
- `==` compares list contents. `is` checks whether the objects are identical. Two separate lists can have equal contents.
- `id()` gives an identity number for an object during its lifetime. Thinking in terms of identity is enough here; its relationship to a physical memory address is an implementation detail.

Inside that list, each numbered slot also refers to an element. Replacing a slot changes which element it refers to. Adding an element adds another slot/reference; it does not create a new variable name for you.



````python
items = ['coffee', 'bagel', 'juice', 'muffin']
receipt = items

print(id(items))        # some large number, like 4395782592
print(id(receipt))      # the exact same number
print(items is receipt) # True: `is` asks whether two variables reference the same object
````

The **id()** function returns a number that identifies where a value lives. Because `items` and `receipt` hold the same reference, they are two names for one list, not two separate lists. Two variables that reference the same list are called **aliases**.

## Adding to the Front vs the Back

The row-of-references idea also explains why some list operations take longer than others. `append()` writes a reference into the empty slot at the end of the row, and nothing else moves. `insert(0,value)` has to make room at the front of the list so every reference already in the row has to shift one slot to the right before adding in the new reference.

This program adds 100000 numbers to the back of one list and to the front of another. time.perf_counter() returns the current time in seconds, so subtracting the start time from the end time measures how long each loop took:

````python
import time

start = time.perf_counter()
back = []
for number in range(100000):
    back.append(number)
print(f"append took {time.perf_counter() - start:.3f} seconds")

start = time.perf_counter()
front = []
for number in range(100000):
    front.insert(0, number)
print(f"insert(0) took {time.perf_counter() - start:.3f} seconds")
````

### Tutor note: what the timing shows

`insert(0)` comes out many times slower than `append`. The exact times depend on the computer.

The reason is the row of references. With 100,000 elements, every `insert(0, ...)` shifts every existing reference one slot to the right. `append` never shifts anything.

`pop(0)` has the same problem in reverse. Removing from the front shifts every remaining reference one slot left, while `pop()` from the end shifts nothing.

| Fast (nothing shifts) | Slow on big lists (everything shifts) |
|---|---|
| `append(value)` | `insert(0, value)` |
| `pop()` | `pop(0)` |

This doesn't matter for a receipt with 5 items. It matters for big lists, and that's why most programs build lists with `append`.

## Passing a List to a Function

````python
def clear_receipt(receipt_items):
    # 3. Upon receiving the reference, this function modifies the list
    receipt_items.clear()

# 1. The list below is created in the heap and the "reference" to its location is stored in items
items = ['coffee', 'bagel', 'juice', 'muffin']

# 2. When we invoke clear_receipt, we assign the "reference" to the function parameter "receipt_items"
clear_receipt(items)

# 4. The list referenced by items has been modified by the function!
print(items)  # Prints []
````

A function whose job is to change a list can change the caller's list directly. Because the caller and the function share one list, the function does not need to build a new list and return it, and every part of the program that references that list can see the change. 

>A function can never change a caller's string, number, or boolean this way. The parameter still receives a reference, but an immutable value has no method or bracket assignment that changes it in place.

You can still reassign a variable to change what value it stores but you cannot mutate the value itself.

````python
def add_one_and_print(x):
    x += 1 # shorthand for x = x + 1
    print(x)

my_num = 10
add_one_and_print(my_num)  # 11
print(my_num) # 10
````

`x += 1` looks like it changes the value in `x`, but the number 10 cannot change. Python computes the new value 11 and reassigns `x` to reference it. `my_num` still references `10`.

### Tutor note: mutating a parameter vs reassigning a parameter

Section 4 comes down to one rule:
- **Mutating** a parameter (`.clear()`, `.append()`, `items[0] = ...`) changes the object the caller also references, so the caller sees the change.
- **Reassigning** a parameter (`x = ...`, `x += 1` on a number) only moves the function's own label. The caller's variable still references the old object.

```python
def reset(receipt_items):
    receipt_items = []        # reassigns the parameter only

items = ['coffee', 'bagel']
reset(items)
print(items)                  # ['coffee', 'bagel']: unchanged
```

```text
before:  items ──► ['coffee', 'bagel'] ◄── receipt_items
after:   items ──► ['coffee', 'bagel']     receipt_items ──► []
```

Compare `clear_receipt` above. `.clear()` went through the shared reference and emptied the caller's list. `receipt_items = []` just pointed the parameter at a new empty list, and that list disappears when the function ends.

# 5. Copying a List Before Changing It

A store is running a sale that takes $1 off every item. The receipt should show each item's regular price next to its sale price. This calculates the sale prices:

````python
def apply_discount(prices):
    for i in range(len(prices)):
        prices[i] = prices[i] - 1
    return prices

prices = [3.50, 2.25, 4.00, 3.00]
sale_prices = apply_discount(prices)

print(prices)       # [2.5, 1.25, 3.0, 2.0]
print(sale_prices)  # [2.5, 1.25, 3.0, 2.0]
````

This prints the prices and sale_prices. The sale prices are correct but the regular prices are gone. The parameter is referencing the same list as the variable outside of it. So any changes to prices inside the function are made to the same list. Here it changed every slot of the list that both prices and the parameter reference. 

To leave the caller's list alone, we can first copy the list and change only the copy. There are two ways of doing this: `x = list(list_name)` and `x = list_name[:]`.

````python
prices = [3.50, 2.25, 4.00, 3.00]

# Two ways to copy:
copy1 = list(prices)   # the list() function builds a new list from any sequence
copy2 = prices[:]      # a slice with no start or end creates a copy of the whole thing
````

With the copy `apply_discount` changes only its own list and returns it, leaving the original list untouched.

````python
def apply_discount(prices):
    sale_prices = list(prices)
    for i in range(len(sale_prices)):
        sale_prices[i] = sale_prices[i] - 1
    return sale_prices

prices = [3.50, 2.25, 4.00, 3.00]
sale_prices = apply_discount(prices)

print(prices)       # [3.5, 2.25, 4.0, 3.0]
print(sale_prices)  # [2.5, 1.25, 3.0, 2.0]
````

`[*list_name]` is a third way, but it uses unpacking, which is covered in Section 6.

## Copying List Challenge

This function changes the list it is given. Rewrite it so that the caller's list stays the same!

```py
def shorten(items):
    items.pop()
    return items

items = ['coffee', 'bagel', 'juice', 'muffin']
shorter_items = shorten(items)

print(items)          # should print ['coffee', 'bagel', 'juice', 'muffin']
print(shorter_items)  # should print ['coffee', 'bagel', 'juice']
```

````python
def shorten(items):
    short_items = list(items)
    short_items.pop()
    return short_items

items = ['coffee', 'bagel', 'juice', 'muffin']
shorter_items = shorten(items)

print(items)          # should print ['coffee', 'bagel', 'juice', 'muffin']
print(shorter_items)  # should print ['coffee', 'bagel', 'juice']
````

### Tutor note: the copying challenge

This solution copies first and then uses `pop()` on the copy. That's the pattern from this section.

Slicing also makes a new list, so it works too: `return items[:-1]` returns everything except the last element in one line. The `-1` stop index means "stop before the last element." The caller's list is never touched, because slicing only reads.

# 6. Advanced List Syntax

## Unpacking

`split()` breaks a string into a list of pieces. **Unpacking** makes it possible to take the values from a list and assign them to distinct variables in one statement.

The syntax puts several variable names, separated by commas, on the left side of the assignment operator = and the list on the right side:

````python
line = "coffee: 3.50"
item, price = line.split(": ")

print(item)   # coffee
print(price)  # 3.50
````

### Tutor note: unpacking gives you strings

`split()` always produces a list of **strings**, so `price` here is `'3.50'`, not the number `3.5`. The comment `# 3.50` hides this, because `print` shows strings without quotes.

- `type(price)` is `<class 'str'>`.
- `price + 1` raises `TypeError: can only concatenate str (not "int") to str`.
- To use it as a number, cast it first (chapter 1.5): `price = float(price)`.

Number of variables has to match the number of elements. If the numbers do not match, there's a `ValueError`:`too many values to unpack` when the list has more elements than variables, and `not enough values to unpack` when it has fewer. The exception is a variable with a `*` in front, the variable collects the rest as a list:

````python
items = ['coffee', 'bagel', 'juice', 'muffin']

first_item, *other_items = items

print(first_item)   # coffee
print(other_items)  # ['bagel', 'juice', 'muffin']
````

### Tutor note: the starred variable is always a list

The `*` variable collects **whatever is left** into a list, even if that's one element or none:

| Code | `first` | `rest` |
|---|---|---|
| `first, *rest = ['tea', 'scone', 'jam']` | `'tea'` | `['scone', 'jam']` |
| `first, *rest = ['tea', 'scone']` | `'tea'` | `['scone']` |
| `first, *rest = ['tea']` | `'tea'` | `[]` |

The star can go anywhere, but only on one variable: `*rest, last = items` puts the last element in `last` and everything before it in `rest`.

The `*` also works inside square brackets. There, it unpacks the elements of one list into a new list, which makes a copy of the list. You can even add extra elements while you copy:

````python
items = ['coffee', 'bagel', 'juice', 'muffin']
copy = [*items]

def add_item(items, item):
    new_items = [*items, item]
    return new_items

more_items = add_item(items, 'cookie')

print(items)       # Prints ['coffee', 'bagel', 'juice', 'muffin']
print(more_items)  # Prints ['coffee', 'bagel', 'juice', 'muffin', 'cookie']
````

### Tutor note: three ways to copy, and two ways to add

You now know three ways to copy a list: `list(items)`, `items[:]`, and `[*items]`. They all make a new list with the same elements.

Look at `add_item` above. It adds `'cookie'` without changing the caller's list. Compare that with `items.append('cookie')`:

| | `[*items, item]` | `items.append(item)` |
|---|---|---|
| Makes a new list? | Yes | No |
| Changes the original? | No | Yes |
| Returns | The new list | `None` |

The `*` can go anywhere in the brackets, so `['tea', *items]` builds a new list with `'tea'` at the **front**. It's the non-mutating version of `items.insert(0, 'tea')`.

## Looping with `enumerate()`

When we need the position of each element as well as the element, we can use `enumerate()`. On each pass it produces a pair, an index and an element, and unpacking splits the pair into two variables.

````python
items = ['coffee', 'bagel', 'juice', 'muffin']
prices = [3.50, 2.25, 4.00, 3.00]

for index, item in enumerate(items):
    print(f"{item}: ${prices[index]:.2f}")

# coffee: $3.50
# bagel: $2.25
# ...
````

### Tutor note: three ways to loop over a list

| Loop | Gives you | Use it when |
|---|---|---|
| `for item in items:` | Each element | You only need the values |
| `for i in range(len(items)):` | Each index | You need positions, and use `items[i]` to get values |
| `for i, item in enumerate(items):` | Both | You need the value **and** its position |

The receipt from Section 2 used `range(len(items))`. This `enumerate()` version does the same job but reads more clearly, because `item` already holds the current element.

**Numbering from 1:** `enumerate()` takes an optional `start` argument. `enumerate(items, start=1)` gives `(1, 'coffee')`, `(2, 'bagel')`, and so on, which is how people expect numbered lists to look. Only the number changes. If that number is used to index a parallel list, it will be off by one.

## 2D Lists

A list can contain other lists. When the inner lists contain values, it's a 2D list or a `"matrix"`.

The receipt can be stored as one 2D list instead of two parallel lists.

````python
receipt = [
    ['coffee', 3.50],
    ['bagel', 2.25],
    ['juice', 4.00],
    ['muffin', 3.00],
]

print(receipt[1])      # ['bagel', 2.25]
print(receipt[1][0])   # bagel
print(receipt[1][1])   # 2.25

print(len(receipt))     # 4, the number of rows
print(len(receipt[0]))  # 2, the number of columns in the first row
````

The first index picks the "row" and the second index picks the "column". Each row is a list, so Python reads `receipt[1][1]` in two steps: `receipt[1]` produces the row `['bagel',2.25]` and `[1]` then picks index 1 of that row, which is `2.25`.

### Tutor note: 2D lists vs parallel lists

With parallel lists, the name and price of one purchase live in two different lists. You have to keep them lined up yourself, and removing one without the other breaks every pair after it. In a 2D list, each row keeps its name and price together, so they can't get out of step.

**Looping over rows:** each row is a two-element list, so unpacking works right in the loop line:

```python
for name, price in receipt:
    print(f"{name}: ${price:.2f}")
```

**The trade-off:** `receipt[1][1]` doesn't say what it means. You have to remember that column 1 is the price. Dictionaries (1.8) fix that by using names instead of numbers.

**Shallow copies:** `list(receipt)` makes a new **outer** list, but the rows inside are not copied. Both lists reference the same row lists, the same row-of-references picture one level deeper:

```text
receipt ──► [ row 0 | row 1 ]
               │       │
               ▼       ▼
     ['coffee', 3.5]  ['bagel', 2.25]
               ▲       ▲
               │       │
copy    ──► [ row 0 | row 1 ]
```

So after `copy = list(receipt)`, running `copy[0][1] = 9.99` changes `receipt[0]` too. Replacing a whole row, as in `copy[0] = ['tea', 1.75]`, only affects `copy`.

## Tuples: Lists That Cannot Change

A **tuple** is a list that cannot be changed after it is created. It's written with parentheses `()` instead of square brackets. Everything that reads a list works on it: indexing, slicing, `len()`, `in`, and `for`.

````python
line = ('coffee', 3.50)

print(line[0])          # coffee
print(len(line))        # 2
print(3.50 in line)     # True

line[1] = 3.00          # TypeError: 'tuple' object does not support item assignment
````

We use tuples when values belong together and should never be added to or removed from. Python itself hands back tuples: `enumerate()` produces `(index, value)` pairs, and a dictionary's `.items()` produces `(key, value)` pairs. When we see `choice in ("1", "2")`, that is a tuple being used as a small fixed set of options.

### Tutor note: when to choose a tuple

Use a **list** when the collection will grow, shrink, or change, like a cart that gets items added. Use a **tuple** when the values are a fixed group that belongs together, like one `(name, price)` pair or a fixed set of menu options. If you try to change a tuple, the `TypeError` tells you that you were about to change something that shouldn't change.

The cell above ends with an intentional `TypeError`, like the string example in Section 3.

**A classic trap:** the parentheses don't make a tuple; the comma does.

| Code | Type |
|---|---|
| `('coffee')` | `str` (the parentheses are just grouping, like in math) |
| `('coffee',)` | `tuple` with one element |
| `'coffee', 3.50` | `tuple`, even without parentheses |

**Returning several values:** `return days, tips` returns one tuple, `(days, tips)`. The caller unpacks it with `days, tips = parse_shifts(week)`, the same unpacking as `item, price = line.split(": ")`.

# 7. Summary

- A **list** stores many values, in order, inside one value. Use one when a program needs to **remember every value**, not just a running total. Start with `[]` and build it with `.append()`.
- Lists work like strings for reading: indexes start at `0`, `-1` is the last element, slicing returns a **new** list and stops **before** the end index, `len()` counts elements, and `in` checks for a whole matching element.
- Unlike strings, lists are **mutable**. Bracket assignment replaces an element, `.append()` / `.insert()` add, and `.pop()` / `.remove()` / `.clear()` remove. These methods change the list in place. Most return `None`; `.pop()` returns what it removed.
- A variable holds a **reference** to an object, not the object itself. Assigning a list to another variable creates an **alias** (two names, one list), not a copy. `is` checks for the same object, and `==` checks for the same contents.
- Passing a list to a function passes a reference, so **mutations inside the function are visible to the caller**. Reassigning a parameter only moves the function's own label. Numbers, strings, and booleans are immutable, so a function can never change the caller's value.
- To change a list without affecting the caller, **copy it first** with `list(items)`, `items[:]`, or `[*items]`. These are **shallow** copies: lists inside the list are still shared.
- `.append()` is fast. `.insert(0, ...)` gets slow on big lists because every reference has to shift.
- **Unpacking** assigns a list's elements to several variables at once. The counts must match, unless a `*` variable collects the rest as a list.
- **`enumerate()`** gives `(index, element)` pairs, for when you need both.
- A **2D list** is a list of lists, read with `[row][column]`. It keeps related values together instead of spread across parallel lists.
- A **tuple** is an immutable sequence written with `()`. Use it for fixed groups of values. `enumerate()` pairs and functions that return several values give tuples.

# 8. Quick Reference

**Reading a list** (`items = ['coffee', 'bagel', 'juice']`)

| Code | Result | Notes |
|---|---|---|
| `items[0]` | `'coffee'` | First element |
| `items[-1]` | `'juice'` | Last element |
| `items[0:2]` | `['coffee', 'bagel']` | New list; stop index not included |
| `items[:]` | `['coffee', 'bagel', 'juice']` | Copy of the whole list |
| `len(items)` | `3` | Number of elements |
| `'bagel' in items` | `True` | Checks for a whole equal element |
| `'bag' in items` | `False` | No partial matches, unlike strings |

**Changing a list** (all of these mutate the original)

| Code | What it does | Returns |
|---|---|---|
| `items[i] = value` | Replaces the element at index `i` | — |
| `items.append(value)` | Adds to the end | `None` |
| `items.insert(i, value)` | Adds at index `i`, shifting the rest right | `None` |
| `items.pop()` | Removes the last element | The removed element |
| `items.pop(i)` | Removes the element at index `i` | The removed element |
| `items.remove(value)` | Removes the first equal element | `None` |
| `items.clear()` | Removes every element | `None` |

- Never write `items = items.append(x)`. `append` returns `None`, so this sets `items` to `None`.

**Common errors**

| Error | Usually means |
|---|---|
| `IndexError` | Index past the end of the list, or `pop()` on an empty list |
| `ValueError` | `remove()` a value that isn't there, or unpacking with the wrong number of variables |
| `TypeError: ... does not support item assignment` | Tried to change a string or tuple in place |

**Aliases vs copies**

```python
alias = items         # same list, two names
copy1 = list(items)   # new list
copy2 = items[:]      # new list
copy3 = [*items]      # new list

alias is items        # True:  same object
copy1 is items        # False: different objects
copy1 == items        # True:  same contents
```

- A function that **mutates** a list parameter changes the caller's list. Copy first if the caller's list must stay the same.

**Unpacking**

```python
item, price = "coffee: 3.50".split(": ")   # item = 'coffee', price = '3.50' (a string!)
first, *rest = items                       # rest is always a list
new_items = [*items, 'cookie']             # new list, items unchanged
```

**Looping over a list**

| Loop | Gives you |
|---|---|
| `for item in items:` | Each element |
| `for i in range(len(items)):` | Each index |
| `for i, item in enumerate(items):` | Index and element |

**2D lists**

```python
receipt = [['coffee', 3.50], ['bagel', 2.25]]
receipt[1]       # ['bagel', 2.25]  (row)
receipt[1][0]    # 'bagel'          (row 1, column 0)
len(receipt)     # 2  (number of rows)
```

**Lists vs tuples vs strings**

| | List | Tuple | String |
|---|---|---|---|
| Written as | `['a', 'b']` | `('a', 'b')` | `'ab'` |
| Mutable? | Yes | No | No |
| Index, slice, `len`, `in`, `for` | Yes | Yes | Yes |
| Use for | Collections that change | Fixed groups of values | Text |
