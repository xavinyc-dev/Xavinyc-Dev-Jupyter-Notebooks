# 1. Key Terms

| **Term** | **Definition** |
|---|---|
| **Loop** | A way to make Python repeat code multiple times, which makes repetitive tasks easy. |
| **Iteration** | One pass through a loop. **Iterating** over something means going through it one element at a time. |
| **Iterable** | Any value a `for` loop can go through one element at a time. A `range()` gives one number at a time and a string gives one character at a time, so they are both iterables. Whatever comes after `in` in a `for` loop is usually an iterable. |
| **Element** | One value inside a list (or one character inside a string). `[3, 7, 12]` has 3 elements. |
| **`for` loop** | Repeats its code block once for each element of an iterable. Used for repetitive tasks done a known number of times, like counting to 10. |
| **`range()`** | A built-in function that gives an iterable of numbers: `range(5)` gives `0, 1, 2, 3, 4`. Used with a `for` loop to repeat something a set number of times. |
| **`while` loop** | Repeats its code block as long as its condition is `True`, so it can run an unknown number of times. It stops once the condition becomes `False`. |
| **Infinite loop** | A loop whose condition never becomes `False`, so it never ends on its own. `while True` makes one on purpose, and `break` is how to get out of it. |
| **`break`** | Ends a loop immediately. The program continues with the first line after the loop. |
| **`continue`** | Skips the rest of the current pass and jumps back to the top of the loop for the next pass. |
| **REPL** | Short for Read-Evaluate-Print Loop: a program that keeps reading input, doing something with it, and printing a result until the user quits. |
| **Running total** | A variable created before a loop, like `total = 0`, that gets added to on every pass. |
| **Nested loops** | Loops inside of other loops. The inner loop runs all the way through on every pass of the outer loop. Useful for grids, rows and columns, tables, and every combination of two lists. |
| **Loop challenges** | Practice problems that use loops to solve real tasks, such as counting results or repeating until something happens. |
| **List** | An ordered collection of values inside square brackets, like `[3, 7, 12]`. A list is an iterable, so a `for` loop can go through it one element at a time. |
| **`sum()`** | A built-in function that adds every number in a list and returns the total: `sum([3, 7, 12])` returns `22`. |
| **`.append()`** | A list method that adds one element to the end of a list: `numbers.append(5)`. |
| **List comprehension** | A one-line `for` loop inside `[]` that builds a new list, like `[int(v) for v in values]`. |

# 2. Intro to Iteration

**Iteration** means doing the same steps over and over, where each round gets closer to the finish. Each single round is called an **iteration**: one pass through a loop.

Picture this: someone asks you to count the vowels in a word. To do it by hand you would:

1. Keep a tally in your head, starting at 0
2. Go through each character in the word, one at a time
    - If it's a vowel, add 1 to the tally
    - If you've reached the end of the word, stop

With the string tools from 1.5 we could write a program that does this, checking one letter at a time with an `if` statement:

````python
word = input("Give me a word: ").strip().lower()

vowels = 0
if word[0] in "aeiou":
    vowels += 1
if word[1] in "aeiou":
    vowels += 1
if word[2] in "aeiou":
    vowels += 1
if word[3] in "aeiou":
    vowels += 1

print(f"{word} has {vowels} vowels.")
````

This program has two problems, and both come from the same cause: **the program can't know ahead of time how long a word the user will type.**

- If the user types `ada`, the program crashes with an `IndexError` at `word[3]`. A 3 letter word only has indexes `0`, `1`, and `2`, so there is no index `3` (the same crash from Parsing Input in 1.5).
- If the user types `lovelace`, the program only checks the first 4 letters (`l`, `o`, `v`, `e`), so it says 2 vowels when the word really has 4.

Adding more `if` statements doesn't fix it, because every length of word would need a different number of them. What we need is a way to say "do this check once for every letter, however many letters there are." That's what a loop does.

# 3. The `for` loop

A `for` loop runs its code block once for each **element** in an **iterable** value, like a string. The elements of a string are its characters, so a `for` loop over a string runs once for each character:

````python
for letter in "hello":
    print(letter)

# h
# e
# l
# l
# o
````

Each time through the loop, Python does these steps in order:

1. Take the next element from the iterable and assign it to the variable (`letter`)
2. Execute the code block
3. If the iterable has more elements, go back to step 1
4. Otherwise, the loop ends and the program continues with the first line after the loop

| **Loop #** | **1. Assign the next element** | **2. Run the code block** | **3. More elements?** |
|---|---|---|---|
| 1 | `letter = "h"` | `print("h")` | Yes, go to loop 2 |
| 2 | `letter = "e"` | `print("e")` | Yes, go to loop 3 |
| 3 | `letter = "l"` | `print("l")` | Yes, go to loop 4 |
| 4 | `letter = "l"` | `print("l")` | Yes, go to loop 5 |
| 5 | `letter = "o"` | `print("o")` | No, the loop is done |

>**Note:** Anything a `for` loop can go through one element at a time is called an **iterable**, and going through its elements one at a time is called **iterating** over it. A string is an iterable of characters, `range()` gives an iterable of numbers, and lists are iterables too.

Every `for` loop has the same form, no matter what it iterates over:

```py
for variable_name in iterable:
    # code block
```

- `for` and `in` are keywords that every `for` loop uses.
- `variable_name` is a variable the loop creates. We choose the name, just like any other variable.
- `iterable` is the value the loop goes through, like a string, a list, or a `range()`.
- The colon `:` and the indented code block work the same way they do in an `if` or `def` statement.

Predict, then run. The variable names in these loops don't match what the loops go through. Does each loop run? What does each one print?

````python
for number in "abc":
    print(number)

for letter in "123":
    print(letter)
````

**My prediction:** Both loops run and print each character of the string, one at a time, no matter what the variable is named.

Output:
```
a
b
c
1
2
3
```

Both loops run because Python doesn't read the meaning of a variable name. `number` and `letter` are just labels we picked.

- `"abc"` is a string, so on each pass `number` holds the next character: `"a"`, then `"b"`, then `"c"`.
- `"123"` is also a string, so `letter` holds the **strings** `"1"`, `"2"`, and `"3"`, not numbers.

Since Python never checks if a name makes sense, the name is there for the people reading the code. Pick a name that describes what the variable will hold, like `letter` for the characters of a word, so the next person reading it (including us later) can tell what it holds.

## Looping Over a String

With a `for` loop the vowel counter checks every letter of the word, however many letters there are:

````python
word = input("Give me a word: ").strip().lower()

vowels = 0
for letter in word:
    if letter in "aeiou":
        vowels += 1

print(f"{word} has {vowels} vowels.")
````

The loop runs once for each character, so `xavier` gets 6 checks and `rodriguez` gets 9, and the program never asks for an index that doesn't exist. Both problems from the intro are fixed.

`vowels = 0` is created **before** the loop so the count keeps going between passes. If it were inside the loop it would reset to `0` on every letter.

`.strip()` and `.lower()` clean the input first, so `letter in "aeiou"` also counts capital vowels like the `A` in `Ada`.

## Looping a Number of Times with `range()`

Not every repetition has a string to go through. Sometimes we just need to repeat something a certain number of times.

Say you are given the task of rolling a 6-sided die 100 times and documenting the result of each.

```
1. a 4 was rolled
2. a 2 was rolled
3. a 6 was rolled
4. a 6 was rolled
5. a 1 was rolled
...
```

1. Gather your materials (pen, paper, a die)

2. Until you've rolled 100 times you will:
    
    a. roll the die

    b. record the result by writing: "[roll number]. a [roll result] was rolled"

3. Make sure to increase the roll number after each roll.

````python
import random

print(f"1. a {random.randint(1, 6)} was rolled")
print(f"2. a {random.randint(1, 6)} was rolled")
print(f"3. a {random.randint(1, 6)} was rolled")
print(f"4. a {random.randint(1, 6)} was rolled")
print(f"5. a {random.randint(1, 6)} was rolled")
# and so on until you reach 100! 🫠
````

The first line, `import random`, loads a tool that comes with Python, and `random.randint(1, 6)` returns a random integer between 1 and 6 (including 1 and 6). Chapter 1.9 explains how importing works, but for now just focus on what `random.randint(1, 6)` does. Writing it out line by line 100 times would be a lot, especially since we'd have to change the roll number on every line.

This time there's no string to iterate over, only a number of times to repeat. The `range(stop)` function fills that gap. It gives us an iterable of numbers starting at `0` and ending **before** the `stop` number: `range(100)` gives `0`, `1`, `2`, ... up to `99`. That's 100 numbers, so a `for` loop over `range(100)` runs 100 times.

```py
for i in range(how_many_times):
    # do something
```

Predict, then run. What is the first number assigned to `roll_number`, and what is the last number?

````python
import random

for roll_number in range(20):
    print(f"{roll_number + 1}. a {random.randint(1, 6)} was rolled")
````

`range(20)` produces the numbers `0` to `19`, and `roll_number` is assigned each number one at a time, one number per loop. 20 numbers, 20 values for `roll_number`, 20 rolls. (Change it to `range(100)` for the full 100 rolls.)

1. Take the next number from the range and assign it to `roll_number`
2. Execute the code block
3. If the range has more numbers, go to step 1
4. Otherwise, the loop ends and the program continues with the first line after the loop

The first value `roll_number` gets is `0` and the last is `19`, not `20`! That's why we print `roll_number + 1`, so the rolls start at `1` and end at `20` in the output. Counting from zero is the normal way in Python, just like string indexes start at `0`.

## Changing the Starting Point for `range()`

`range()` can also take a starting point and a step size: `range(start, stop, step)`. The `stop` number is always left out of the range, the same way a slice stops before its `end`.

````python
import random

for i in range(1, 21):      # 1, 2, 3, ... 20
    print(f"{i}. a {random.randint(1, 6)} was rolled")

for i in range(0, 20, 5):   # 0, 5, 10, 15
    print(i)

for i in range(10, 0, -1):  # 10, 9, 8, ... 1
    print(i)
````

- `range(1, 21)` starts at `1` instead of `0`, so we don't need the `+ 1` anymore. It stops before `21`, so the last number is `20`.
- `range(0, 20, 5)` counts by 5s: the third number is the **step**, how much to add each time.
- `range(10, 0, -1)` counts backwards. A negative step subtracts each time, and it stops before `0`, so the last number is `1`.

Because `range()` gives numbers, it can also give the indexes of a string. That's one of the main reasons `range()` starts counting at `0`!

Predict, then run. This loop goes over a `range()` instead of the string itself. It uses the variable `i` (short for "index") to hold each number in `range(len(word))`.

````python
word = "loop"
for i in range(len(word)):
    print(i, word[i])
````

Output:
```
0 l
1 o
2 o
3 p
```

- `len(word)` is `4`, so `range(len(word))` gives `0`, `1`, `2`, and `3`, which are exactly the indexes of `word`.
- On each pass `i` holds one index, and `word[i]` uses bracket notation to get the character at that index.

Looping over `range(len(word))` gives us each character's position **and** the character. When we only need the characters, `for letter in word` is shorter and there's no index to get wrong.

## Looping Over a List

A **list** is an ordered collection of values written inside square brackets `[]`, with a comma between each value. Each value is called an **element**. Chapter 1.7 covers lists in depth, but we only need the basics to loop over one.

```py
prices = [3, 7, 12]            # a list of integers
words = ["hay", "needle"]      # a list of strings
mixed = [9, "7", True]         # a list can hold different data types
```

A list is an **iterable**, so a `for` loop goes through it one element at a time, the same way it goes through the characters of a string.

Predict, then run. What does this loop print?

````python
prices = [3, 7, 12]

for price in prices:
    print(price * 2)
````

Output:
```
6
14
24
```

On each pass `price` holds the next **element** of the list: `3`, then `7`, then `12`. The loop runs 3 times because the list has 3 elements.

## Adding Up a List with `sum()`

To add up a list, we use the same pattern as the vowel counter: start a **running total** before the loop, then add to it on every pass.

````python
prices = [3, 7, 12]

total = 0
for price in prices:
    total += price

print(total)  # 22
````

| **Loop #** | **`price`** | **`total += price`** |
|---|---|---|
| 1 | `3` | `0 + 3 = 3` |
| 2 | `7` | `3 + 7 = 10` |
| 3 | `12` | `10 + 12 = 22` |

`total = 0` goes **before** the loop for the same reason `vowels = 0` does: if it was inside the loop, it would reset to `0` on every pass.

Adding up a list is so common that Python has a built-in function for it. `sum()` takes a list and returns the total, so it does the whole loop above in one line:

````python
prices = [3, 7, 12]

print(sum(prices))  # 22
````

>**Note:** `sum()` only works when **every** element is a number. A single string in the list crashes it with a `TypeError`, because Python can't add an `int` and a `str`:
>
>```py
>sum([5, "10"])  # TypeError: unsupported operand type(s) for +: 'int' and 'str'
>```

## Building a List with `.append()`

`.append()` is a list **method** that adds one element to the **end** of a list. We call it with a dot after the list's name, the same way we call string methods like `.strip()` and `.lower()` from 1.5.

```py
list_name.append(element)
```

Predict, then run. What does `numbers` hold at the end?

````python
numbers = []          # an empty list, no elements yet

numbers.append(5)
numbers.append(10)
numbers.append(15)

print(numbers)
````

Output:
```
[5, 10, 15]
```

Each `.append()` adds one element to the end, so the list grows in the order we added them.

>**Note:** `.append()` is different from string methods in one big way: it **changes the list itself** and returns `None`. String methods can't change a string (strings are immutable), so they return a new string that we have to save. That means `numbers = numbers.append(5)` is a mistake: it saves the returned `None` into `numbers` and our list is gone. Just write `numbers.append(5)` on its own line.

`.append()` is most useful inside a loop. This is the **build-a-list pattern**:

1. Create an empty list **before** the loop (just like `total = 0`, so it doesn't reset on every pass)
2. `.append()` one element **inside** the loop, on every pass
3. Use the finished list **after** the loop

Predict, then run. What gets printed?

````python
evens = []

for i in range(2, 11, 2):
    evens.append(i)

print(evens)
````

Output:
```
[2, 4, 6, 8, 10]
```

| **Loop #** | **`i`** | **`evens` after `.append(i)`** |
|---|---|---|
| 1 | `2` | `[2]` |
| 2 | `4` | `[2, 4]` |
| 3 | `6` | `[2, 4, 6]` |
| 4 | `8` | `[2, 4, 6, 8]` |
| 5 | `10` | `[2, 4, 6, 8, 10]` |

Inside a function, we `return` the list **after** the loop ends. Watch the indentation: `return` lines up with `for`, not with the code inside the loop.

````python
def squares_up_to(n):
    squares = []
    for i in range(1, n + 1):
        squares.append(i * i)
    return squares            # after the loop, so it runs once at the end

print(squares_up_to(4))   # [1, 4, 9, 16]
````

If `return squares` was indented inside the loop, the function would return on the very first pass and `squares_up_to(4)` would give `[1]`. A `return` ends the function right away (see Return Statements in 1.3).

>**Note:** `.append()` adds an element to a list. `assert` is a different thing: it's a statement that only **checks** if something is `True`. It can't add anything to a list, and `numbers.assert(i)` is a `SyntaxError`.

## Converting Every Element with `int()`

Values that come from `input()` are always strings, so a list of them can't be added up yet. `int()` converts a value to an integer, but it only takes **one** value at a time. There is no `int()` for a whole list:

```py
int("7")         # 7
int(9)           # 9 (already an integer, so int() just gives it back)
int(["7", "3"])  # TypeError: int() argument must be a string, a bytes-like object or a real number, not 'list'
```

`int()` also doesn't change the variable we give it. It **returns** a new value (see Return Statements in 1.3), so we have to save that value in a variable.

Predict, then run. What type is `x` after each `print()`?

````python
x = "7"
int(x)
print(x, type(x))

x = int(x)
print(x, type(x))
````

Output:
```
7 <class 'str'>
7 <class 'int'>
```

- On line 2, `int(x)` returns the integer `7`, but nothing saves it, so that value is thrown away and `x` is still the string `"7"`.
- On line 5, `x = int(x)` saves the returned value back into `x`, so now `x` is the integer `7`.

To convert a whole list, we use the build-a-list pattern from the last section: loop over the list, run `int()` on each element, and `.append()` every result to a **new** list.

````python
values = ["5", "10", "20"]

numbers = []                      # start with an empty list
for value in values:
    numbers.append(int(value))    # convert the element, then add it to the new list

print(numbers)       # [5, 10, 20]
print(sum(numbers))  # 35
````

`numbers = []` is created before the loop, just like `total = 0`. Each pass converts one element and adds it to the end of `numbers`, so after 3 passes `numbers` holds 3 integers and `sum()` can add them.

## List Comprehensions

Building a new list from an old one with a loop and `.append()` is so common that Python has a shorter way to write it, called a **list comprehension**. It's the same loop, written in one line inside square brackets:

````python
values = ["5", "10", "20"]

numbers = [int(value) for value in values]

print(numbers)  # [5, 10, 20]
````

Read it left to right as "**`int(value)`** for each **`value`** in **`values`**." The part before `for` is what happens to each element, and the part after is the same as the first line of a `for` loop.

| **Loop with `.append()`** | **List comprehension** |
|---|---|
| `numbers = []` | `numbers = [ ... ]` |
| `for value in values:` | `for value in values` |
| `numbers.append(int(value))` | `int(value)` (goes first, before `for`) |

A list comprehension can also skip elements by adding an `if` at the end. Only the elements where the condition is `True` make it into the new list.

Predict, then run. Which scores end up in `valid`?

````python
scores = ["90", "", "85", "n/a", "77"]

valid = [int(score) for score in scores if score.isdigit()]

print(valid)
print(sum(valid))
````

Output:
```
[90, 85, 77]
252
```

`.isdigit()` returns `True` only when a string is all digits, so `""` and `"n/a"` are skipped. Without the `if`, `int("")` would crash with a `ValueError`.

>**Note:** In other people's code we may also see `list(map(int, values))`. It does the same thing as `[int(value) for value in values]`. We use list comprehensions because they read like a `for` loop and can filter with `if`.

## For Loop Challenge

Write a function called `count_heads` that does the following:

1. Takes in a number of flips as a parameter.
2. Flips a coin that many times, keeping count of how many times heads was flipped.
3. Prints back the final number of heads like this:

```
You flipped [heads flipped] heads out of [total] flips! That is [percent]%!
```

>**Note:** A function body can't be empty. Starter code often puts `pass` in a function, a statement that does nothing, to hold its place until we write our own code.

````python
import random

def flip_coin():
    flip_result = random.randint(0, 1)
    return "heads" if flip_result == 1 else "tails"

def count_heads(flips):
    heads = 0
    for flip in range(1, flips + 1):
        if flip_coin() == "heads":
            heads += 1
    return f"You flipped {heads} heads out of {flips} flips! That is {(heads/flips)*100:.2f}%!"

print(count_heads(10))
print(count_heads(100))
````

# 4. While Loops

A `for` loop runs a known number of times. You provide it an iterable, and it runs once for each element until that iterable runs out.

A `while` loop repeats its code block for as long as its condition is `True`. It can be viewed as an `if` statement that repeats itself. One of the most common use-cases for `while` loops is to create a **REPL** program (Read-Evaluate-Print Loop) that keeps prompting the user and that takes in their input until they type a message like `"Exit"`:

````python
user_input = None # user hasn't typed in anything yet
while (user_input != "Exit"):
    user_input = input("type something and hit enter: ")
    print(f"you typed: {user_input}")

print("Good bye!")
````

Before each pass, Python checks the condition for the user input (`user_input != "Exit"`). If the condition is `True`, the block runs again. If the condition becomes `False`, the loop ends and the program continues with the code after. This makes `while` loops the right tool for repeating an unknown number of times.

`user_input = None` is there because the condition is checked **before** the first pass, so `user_input` has to exist already. Without that line, Python stops with `NameError: name 'user_input' is not defined`.

Here's another example: rolling a die until we get a 6. There's no way to know how many rolls it will take, so a `for` loop doesn't fit, but a `while` loop does. `roll = None` gives `roll` a starting value so the condition can be checked before the first roll, and `roll_count` is a counter created before the loop.

````python
import random

roll = None
roll_count = 0
while (roll != 6):
    roll = random.randint(1, 6)
    roll_count += 1

print(f"It took {roll_count} rolls to get a 6!")
````

## Asking Until the Input is Valid

In 1.5, a program could check an age with `isdigit()`, but when the answer wasn't a number, all it could do was print a message and stop. The user had to run the program again. A `while` loop can ask again instead, as many times as it takes, since we don't know how many tries the user will need:

````python
age = input("How old are you? ").strip()

while not age.isdigit():
    print(f"Sorry, {age} is not a number")
    age = input("How old are you? ").strip()

age = int(age)
print(f"Next year you will be {age + 1}")
````

The loop body only runs when `age` fails validation (it isn't all digits). Each pass prints a message and asks again, and it keeps repeating for as long as the answer keeps failing. The program can't know how many wrong answers the user will type, which is why this is a `while` loop. By the time the program reaches `int(age)`, `age` is guaranteed to hold digits, so the cast can't crash.

## Break and Continue

The loop above writes the same `input()` line twice: once before the loop, so `age` holds a value the first time the condition is checked, and once inside the loop to ask again. If the question changes, it has to be changed in two places.

Two keywords let us control a loop from inside its code block:

- `break` ends the loop immediately, like `return` ends a function, and the program continues with the first line after the loop. `break` is how you get out of a loop that is always `True`.
- `continue` skips the rest of the current pass and jumps back to the top of the loop. `continue` keeps the loop going.

With `break` we can ask the question in one place for the validation loop:

````python
while True:
    age = input("How old are you? ").strip()
    if age.isdigit():
        break
    print(f"Sorry, {age} is not a number")

age = int(age)
print(f"Next year you will be {age + 1}")
````

`while True` means the condition is always `True`, so the loop never ends by itself. The only way out is the `break`, which runs only when the user has typed a number. The question is now written in one place.

Here is a program that uses both `break` and `continue`. It adds up the dollar amounts the user types until they type `done`:

````python
total = 0

while True:
    amount = input("Enter an amount in dollars (like 4.50), or done to finish: ").strip().lower()
    if amount == "done":
        break

    digits = amount.replace(".", "", 1)
    if not digits.isdigit():
        print(f"Sorry, {amount} is not a dollar amount")
        continue

    total += float(amount)

print(f"Your total is ${total:.2f}")
````

We don't know how many amounts the user will type, which is why we use a `while` loop. `while True` never ends the loop by itself, so the two statements inside it do the work:

- `break` runs when the user types `done`. The loop ends and the program continues with the `print()` after the loop.
- `continue` runs when the user types something that isn't a dollar amount, like `ten` or nothing at all. It skips the rest of the pass, so `float(amount)` never receives the bad input and can't crash with a `ValueError`. The loop goes straight back to asking for the next amount.

`.isdigit()` returns `False` for `"4.50"` because `.` isn't a digit. `amount.replace(".", "", 1)` removes **one** dot (the `1` limits it to one replacement), so `"4.50"` becomes `"450"` and passes the check, while `"1.2.3"` becomes `"12.3"` and still fails. The new string is saved in `digits`, which is **only used for checking**. Strings are immutable, so `amount` still holds `"4.50"`, and that is what gets added to the total.

`float()` is used instead of `int()` because `int("4.50")` would crash on the decimal, and `{total:.2f}` shows the total with exactly 2 decimal places.

>**Note:** The difference between `break` and `return` is that `return` gives an output back to the program where `break` just ends the loop without outputting anything. `break` also only leaves the loop, while `return` leaves the entire function.

## While Loop Challenge

Write a number guessing game program that:

1. Generates a random number from 1-10. This is the secret number.
2. Asks the user to guess the number using the `input()` function.
3. If the guess is correct, prints a winning message and ends the program.
4. If the guess is incorrect, asks the user to guess again.
5. If the user ever types `"Exit"`, ends the program.

**Bonus features**

1. Keep track of the guesses, and when the guess is correct, print how many guesses it took.
2. Limit the user to 5 guesses. If they guess incorrectly 5 times, they lose! (you'll need to use `break`)

````python
import random


secret_number = random.randint(1, 10)
num_of_guesses = 0
total_guesses = 5

while True:
    guess = (
        input(
            f"Guess the secret number from 1-10! You have {total_guesses} guess{'es' if total_guesses > 1 else ''}: "
        )
        .strip()
        .lower()
    )
    if guess == "exit":
        print("Thanks for playing!")
        break
    if not guess.isdigit():
        print("Thats an invalid digit. Try Again!")
        input("Press Enter to try again...")
        continue
    guess = int(guess)
    num_of_guesses += 1
    total_guesses -= 1
    if total_guesses == 0 and guess != secret_number:
        print("You ran out of guesses!!")
        break
    if guess == secret_number:
        print(
            f"You've Won!! It took {num_of_guesses} guess{'es' if num_of_guesses > 1 else ''}!"
        )
        break
    else:
        print("Wrong! Try Again!")
        continue
````

How my solution works:

- `.strip().lower()` cleans the guess right away, so `Exit`, `EXIT`, and `exit ` with a space all match `"exit"`.
- The `"exit"` check comes **before** `isdigit()`. `"exit"` isn't a digit, so if the digit check ran first it would call `exit` invalid and the user could never quit.
- Invalid input uses `continue` and does **not** count as a guess. Only real numbers add to `num_of_guesses` and take away from `total_guesses`.
- `total_guesses == 0 and guess != secret_number` makes sure a correct guess on the last try still wins instead of counting as a loss.
- `guess{'es' if total_guesses > 1 else ''}` is a conditional expression inside the f-string, so the message says "1 guess" but "2 guesses".

# 5. Nested Loops

A **nested loop** is a loop inside another loop's code block. The loop on the outside is the **outer loop**, and the one inside it is the **inner loop**.

The key rule: **on every pass of the outer loop, the inner loop runs all the way through, from start to finish.** Then the outer loop moves on to its next pass, and the inner loop starts over from the beginning.

```py
for outer_variable in outer_iterable:
    for inner_variable in inner_iterable:
        # runs once for every PAIR of outer and inner values
```



````python
for row in range(1, 4):
    for col in range(1, 4):
        print(f"row {row}, col {col}")
````

The outer loop runs 3 times, and on **each** of those passes the inner loop runs 3 times, so the `print()` runs 3 × 3 = **9** times.

| **Outer pass** | **`row`** | **Inner loop runs with `col` =** | **Lines printed** |
|---|---|---|---|
| 1 | `1` | `1`, `2`, `3` | `row 1, col 1` → `row 1, col 3` |
| 2 | `2` | `1`, `2`, `3` (starts over) | `row 2, col 1` → `row 2, col 3` |
| 3 | `3` | `1`, `2`, `3` (starts over) | `row 3, col 1` → `row 3, col 3` |

`row` stays the same while `col` changes. `row` only moves to its next value after the inner loop has finished completely. It works like a clock: the minute hand (inner loop) goes all the way around before the hour hand (outer loop) moves one step.

>**Note:** The total number of inner passes is **outer count × inner count**. `range(100)` inside `range(100)` is 10,000 passes, so nested loops get slow quickly with big numbers.

## Indentation Decides Which Loop a Line Belongs To

Nested loops are often used to print grids. A line indented under the inner `for` runs on every inner pass, and a line indented under only the outer `for` runs once per outer pass.

````python
for row in range(3):
    for col in range(5):
        print("*", end="")
    print()   # lines up with the inner for, so it runs once per row
````

Output:
```
*****
*****
*****
```

- `print("*", end="")` is inside the inner loop, so it runs 5 times per row. `end=""` (from Output in 1.5) stops `print()` from starting a new line after each star, so the 5 stars land side by side.
- `print()` with nothing inside is lined up with the inner `for`, so it runs once **after** the inner loop finishes. It prints an empty line, which just moves to the next line for the next row.

If `print()` were indented one more level, it would run after every star, and each star would be on its own line. If it were removed, all 15 stars would print on one line.

## A Multiplication Table

A classic use of nested loops is a table: the outer loop picks the row, and the inner loop fills in each column of that row.

````python
for a in range(1, 6):
    for b in range(1, 6):
        print(f"{a * b:>4}", end="")
    print()
````

`{a * b:>4}` is a new f-string format: `>4` means "take up 4 spaces and line the value up on the right." It keeps the columns straight whether the number has 1 digit or 2. Without it, `5` and `25` would take up different widths and the columns would drift.

## Every Combination of Two Lists

Nested loops can go through any iterables, not just `range()`. Looping over one list inside another gives every possible pairing.

Predict, then run. How many shirts are printed?

````python
colors = ["red", "blue"]
sizes = ["S", "M", "L"]

for color in colors:
    for size in sizes:
        print(f"{color} shirt, size {size}")
````

2 colors × 3 sizes = 6 shirts. Every color is paired with every size.

## `break` Only Leaves the Inner Loop

`break` and `continue` only affect the loop they are directly inside. A `break` in the inner loop ends the **inner** loop, and the outer loop keeps going with its next pass.

Predict, then run. Which lines print?

````python
for row in range(1, 4):
    for col in range(1, 4):
        if col == 2:
            break
        print(row, col)
    print(f"done with row {row}")
````

Output:
```
1 1
done with row 1
2 1
done with row 2
3 1
done with row 3
```

On every row, `col == 2` triggers the `break`, so the inner loop stops after `col` is `1`. But the outer loop isn't affected: `done with row ...` still prints, and the next row still starts.

## Where to Put a Counter

The running-total rule still applies, but now there are **two** places "before the loop" could mean:

- Created **before the outer loop**: it counts across the whole program and never resets.
- Created **inside the outer loop, before the inner loop**: it resets at the start of every row, so it counts one row at a time.

Predict, then run. What do `total` and `row_count` print?

````python
total = 0                  # counts everything
for row in range(1, 4):
    row_count = 0          # resets at the start of every row
    for col in range(1, 5):
        total += 1
        row_count += 1
    print(f"row {row}: {row_count}")

print(f"total: {total}")
````

Output:
```
row 1: 4
row 2: 4
row 3: 4
total: 12
```

`row_count` goes back to `0` on every outer pass, so each row reports `4`. `total` was created once, before everything, so it keeps adding: 3 rows × 4 = `12`.

# 6. Summary

- A **loop** repeats a block of code so we don't have to write the same lines over and over, and it works no matter how long the input is.
- A **`for` loop** runs once for each element of an iterable: each character of a string, each number of a `range()`, or each element of a list. Use it when the number of repeats is known.
- **`range(start, stop, step)`** gives numbers to loop over. It starts at `0` by default and always stops **before** `stop`.
- **Lists** are iterables too. `sum()` adds them up, `.append()` builds them one element at a time, and a **list comprehension** builds a new list in one line.
- A **`while` loop** runs as long as its condition is `True`. Use it when the number of repeats is unknown, like asking until the input is valid or until the user types `exit`.
- **`break`** leaves a loop right away, and **`continue`** skips to the next pass. `while True` with a `break` is the common way to write a loop that only ends from the inside.
- **Nested loops** put a loop inside a loop. The inner loop runs all the way through on every pass of the outer loop, which is useful for grids, tables, and every combination of two lists. `break` only leaves the loop it's directly inside.
- Counters, running totals, and empty lists are created **before** the loop, so they don't reset on every pass.

# 7. Quick Reference

**`for` vs `while`**

| | **`for` loop** | **`while` loop** |
|---|---|---|
| Use when | We know how many times (once per element) | We don't know how many times |
| Stops when | The iterable runs out of elements | The condition becomes `False`, or a `break` runs |
| Example | `for letter in word:` | `while not age.isdigit():` |

```python
for variable_name in iterable:
    # runs once per element

while condition:
    # runs as long as condition is True
```

**`range()`**

| Call | Numbers it gives |
|---|---|
| `range(5)` | `0, 1, 2, 3, 4` |
| `range(1, 6)` | `1, 2, 3, 4, 5` |
| `range(0, 20, 5)` | `0, 5, 10, 15` |
| `range(10, 0, -1)` | `10, 9, 8, ... 1` |
| `range(len(word))` | Every index of `word` |

- The `stop` number is **never** included, the same as slicing.

**`break` and `continue`**

| Keyword | What it does | Loop keeps going? |
|---|---|---|
| `break` | Jumps to the first line **after** the loop | No |
| `continue` | Jumps back to the **top** of the loop | Yes |

**Loop patterns**

| Pattern | Before the loop | Inside the loop |
|---|---|---|
| Counter | `count = 0` | `count += 1` |
| Running total | `total = 0` | `total += value` |
| Build a list | `items = []` | `items.append(value)` |

```python
# Ask until the input is valid
while True:
    age = input("How old are you? ").strip()
    if age.isdigit():
        break
    print(f"Sorry, {age} is not a number")

# Keep going until the user quits, skipping bad input
while True:
    answer = input("Number, or done: ").strip().lower()
    if answer == "done":
        break
    if not answer.isdigit():
        continue
    # use the answer here
```

**Nested loops**

```python
for row in range(3):          # outer: 3 passes
    for col in range(4):      # inner: all 4 passes, on EVERY outer pass
        print("*", end="")
    print()                   # once per row, after the inner loop
```

- Total inner passes = outer count × inner count (3 × 4 = 12).
- A counter before the outer loop counts everything. A counter inside the outer loop (before the inner one) resets every row.
- `f"{x:>4}"` lines a value up on the right in 4 spaces, which keeps table columns straight.

**Lists**

| Syntax | What it does | Example |
|---|---|---|
| `[3, 7, 12]` | A list with 3 elements | |
| `sum(lst)` | Adds every number (numbers only) | `sum([3, 7, 12])` → `22` |
| `lst.append(x)` | Adds `x` to the end, changes the list, returns `None` | `[5].append(10)` → list is `[5, 10]` |
| `[int(v) for v in values]` | A new list with every element converted | `["5", "10"]` → `[5, 10]` |
| `[int(v) for v in values if v.isdigit()]` | Converts, and skips elements that fail the `if` | `["5", "", "n/a"]` → `[5]` |

**Common traps**

| Trap | Fix |
|---|---|
| `total = 0` inside the loop resets on every pass | Create it **before** the loop |
| `range(10)` stops at `9` | `range(1, 11)` for `1` to `10`, or print `i + 1` |
| A `while` loop that never changes its variable runs forever | Ask again (or update the variable) inside the loop |
| `numbers = numbers.append(5)` saves `None` | `numbers.append(5)` on its own line |
| `return` indented inside the loop returns after the first pass | Line up `return` with `for` |
| `int(x)` on its own line doesn't change `x` | `x = int(x)` |
| `int(["7", "3"])` crashes with `TypeError` | Convert each element: `[int(v) for v in values]` |
| `sum([5, "10"])` crashes with `TypeError` | Convert every element to a number first |
| `"4.50".isdigit()` is `False` | Check `amount.replace(".", "", 1).isdigit()` instead |
| Printing an error but forgetting `continue`, so the bad input still crashes the next line | `continue` right after the error message |
| `break` in the inner loop doesn't stop the outer loop | `break` only leaves the loop it's directly inside |
| Forgetting `print()` after the inner loop puts the whole grid on one line | Add `print()` lined up with the inner `for` |
| Checking for `"exit"` after `isdigit()`, so it's treated as invalid | Check special words like `"exit"` **before** validating |
