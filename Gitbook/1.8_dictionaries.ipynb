# 1. Key Terms

| **Term** | **Definition** |
|---|---|
| **Dictionary** | A data structure that stores `key: value` pairs, so a program looks each value up by its key instead of by position.<br>• A **key** is the name a value is stored under and looked up by. Each key appears only once in a dictionary.<br>• A **value** is the data stored under a key. It can be any type, including a list or another dictionary. |
| **Record** | A dictionary that describes **one thing** with several labeled values, like `{"item": "coffee", "price": 3.50}`. The programmer picks the keys in advance. |
| **Lookup Table** | A dictionary that describes **many things**, keyed by name: each key is a different item from the data, like `{"coffee": 3.50, "bagel": 2.25}`. New keys get added as the program runs. |
| **List of Dictionaries** | A list that holds many records of the same kind, one dictionary per thing, all with the same keys. |
| **Bracket Notation** | Reads a value by its key: `my_dict["key"]`. If the key is missing it raises a `KeyError`. |
| **`.get()`** | Reads a value by its key but never crashes: a missing key returns `None`, or a default you choose with `my_dict.get("key", default)`. |
| **`in` Operator** | Checks whether a **key** exists: `"price" in purchase`. It does not check values. |
| **Dictionaries are mutable** | You can add, update, or delete key-value pairs after the dictionary is created. |
| **CRUD** | The four things a program does with stored data: **C**reate, **R**ead, **U**pdate, and **D**elete. |
| **Nested Data** | A dictionary or list stored inside another one, such as a receipt dictionary whose `"purchases"` key holds a list of records. |
| **Dictionaries are iterable** | A `for` loop over a dictionary visits its keys. Three methods each return something you can loop over:<br>• **`.keys()`**: all of the keys<br>• **`.values()`**: all of the values<br>• **`.items()`**: every key and its value as `(key, value)` pairs |
| **Reference** | Like lists, dictionaries are held by reference. Copy one with `dict(my_dict)` before changing it inside a function so the caller's dictionary stays the same. |
| **Shallow Copy** | `dict()`, `.copy()`, and `{**my_dict}` make a new outer dictionary, but any list or dictionary nested inside it is still shared with the original. |
| **Hashable** | A key must be hashable, meaning it is an immutable value like a string, number, or tuple. Using a mutable value like a list as a key raises a `TypeError` that says `unhashable type: 'list'`. |

### Tutor note: a few definitions to keep precise

- **Keys are unique, values are not.** Two keys can hold the same value (`{"bagel": 3.00, "muffin": 3.00}` is fine). Assigning to a key that already exists doesn't make a second one, it **replaces** the old value.
- **No positions.** A dictionary has no index numbers. `purchase_1[0]` doesn't give the first pair, it looks for a key that is literally `0` and raises a `KeyError`.
- **Order is remembered.** Dictionaries keep keys in the order they were added, so printing or looping shows them in that order. That's still not the same as having positions you can index.
- **Keys are case-sensitive.** `"Coffee"` and `"coffee"` are two different keys, the same way `"Coffee" == "coffee"` is `False` for strings.
- **Record vs lookup table** is about how you *use* a dictionary, not a different type. Both are `dict`. See the tutor note at the end of Section 4.

# 2. Dictionaries: From Rows to Records

## What is a Dictionary?

**Dictionary** stores each value under a name called a **key** instead of a position, key:value pair. First purchase as a dictionary:

````python
purchase_1 = {"item": "coffee", "price": 3.50}

print(purchase_1["price"])  # 3.5

# Dictionaries can be written in one line or across multiple
purchase_2 = {
    "item": "bagel",
    "price": 2.25
}
print(purchase_2["price"])  # 2.25
````

`{}` surround dictionaries and each entry is a **key:value** pair.
Above `"item"` and `"price"` are the keys written as strings. `"coffee"` and `3.50` are their values which can be any data type.
Keys are usually strings, and the quotes around them are required or else Python reads it as a variable.

Two ways dictionaries are used:

1. `purchase_1` above is an example of a **record**: it keeps pieces of information about one thing. Each key is labeled for a different kind of detail.
2. A **lookup table** takes a name and gives back a matching value. Like a menu, it finds the price of any item from the item's name. The keys are all the same kind of thing (item names), and so are the values (prices).

````python
menu = {
    "coffee": 3.50,
    "bagel": 2.25,
    "juice": 4.00,
    "muffin": 3.00
}
````

### Tutor note: what changes between the two

| | Record (`purchase_1`) | Lookup table (`menu`) |
|---|---|---|
| Describes | **One** thing | **Many** things |
| Keys are | Labels the programmer chose (`"item"`, `"price"`) | The data itself (`"coffee"`, `"bagel"`, ...) |
| Values are | Different kinds of info (a string, a number) | The same kind of info (all prices) |
| Reads like | "This purchase's **price** is 3.50" | "The price **of coffee** is 3.50" |

## Reading Values with Bracket Notation and `.get()`

Bracket notation reads a value by its key.

````python
purchase_1 = {"item": "coffee", "price": 3.50}

print(f"One {purchase_1['item']}, please")  # One coffee, please
````

Predict, then run. There is no "discount" in purchase_1.

````python
print(purchase_1["discount"])
````

Asking a dictionary for a key it does not have is an `KeyError`, like asking a list for an index it does not have.

### Tutor note: the intentional error above

The `KeyError` in the cell above is on purpose, just like the string `TypeError` demo in 1.7. If you use **Run All**, the notebook stops there, so run the cells below it on their own.

The error message names the missing key: `KeyError: 'discount'`. When you see a `KeyError`, the quoted text is the exact key Python looked for and couldn't find, which usually points to a typo or a key that was never added.

`.get()` can be used if we are not sure if the key exists or if we simply just want to search for a key. It returns the value or `None` for a missing key, or a default value we can choose, `.get(key, default_value)`

````python
print(purchase_1.get("discount"))     # None
print(purchase_1.get("discount", 0))  # 0
print(purchase_1.get("item"))         # coffee
````

Default value of `0` is correct here because a purchase with no discount has a discount of `0`.

Can also use `in` to search for the key in the dictionary to avoid the error.

````python
if "discount" in purchase_1:
    print(purchase_1["discount"])
````

### Tutor note: three ways to read a key

| Code | Key exists | Key missing |
|---|---|---|
| `purchase_1["discount"]` | The value | `KeyError` |
| `purchase_1.get("discount")` | The value | `None` |
| `purchase_1.get("discount", 0)` | The value | `0` (the default) |
| `"discount" in purchase_1` | `True` | `False` |

- Use **brackets** when the key should always be there. A crash tells you something went wrong.
- Use **`.get()` with a default** when a missing key has a sensible meaning, like no discount meaning `0`.
- Use **`in`** when the program needs to do something different depending on whether the key exists.

Predict, then run. purchase_1 has a value "coffee". Does the `in` operator tell us if a value is in a dictionary?

````python
if "coffee" in purchase_1:
    print(purchase_1["item"])
````

The `in` operator doesn't check the values, it only checks the keys in the dictionary. `"coffee"` is a value, not a key, so `"coffee" in purchase_1` is `False` and nothing prints.

### Tutor note: checking values instead of keys

To search the values, ask the dictionary for its values first:

```python
"coffee" in purchase_1.values()   # True
"coffee" in purchase_1            # False: only checks keys "item" and "price"
```

This is the opposite of a list. `'coffee' in ['coffee', 'bagel']` checks the elements, which are the values, because a list has no keys.

## Changing a Record

Dictionaries are mutable, like lists. This means that we can perform all four CRUD operations (Create, Read, Update, and Delete), the four things a program does with stored data:

- Create: `my_dict["key"] = value` (bracket notation)

- Read: `my_dict["key"]` (bracket notation)

- Update: `my_dict["key"] = value` (bracket notation)

- Delete: `del my_dict["key"]` or `my_dict.pop("key")`

Create and Update use the same line. Assigning to a key adds a new key-value pair if the key is not in the dictionary yet, and replaces the old value if it is.

````python
purchase_1 = {"item": "coffee", "price": 3.50}

# Bracket notation can be used to modify existing values, or create new ones!
purchase_1["price"] = 3.00
purchase_1["discount"] = 0.50
print(purchase_1)  # {'item': 'coffee', 'price': 3.0, 'discount': 0.5}

# Delete key-value pairs with the `del` keyword followed by bracket notation
del purchase_1["discount"]
print(purchase_1)  # {'item': 'coffee', 'price': 3.0}

# .pop() removes a key-value pair and returns its value
price = purchase_1.pop("price")
print(price)     # 3.0
print(purchase_1)  # {'item': 'coffee'}
````

### Tutor note: `del` vs `.pop()`

| Code | Removes the pair? | Returns | Key missing |
|---|---|---|---|
| `del purchase_1["discount"]` | Yes | Nothing (it's a statement) | `KeyError` |
| `purchase_1.pop("discount")` | Yes | The removed value | `KeyError` |
| `purchase_1.pop("discount", None)` | Yes, if it's there | The value, or `None` | Returns the default, no error |

Use `del` when you just want the pair gone. Use `.pop()` when you also need the value, like saving the price before removing it.

**Compare to lists:** `items.pop()` with no argument removes the last element. A dictionary's `.pop()` always needs a key, because there is no "last position" to default to.

**Updating keeps a key's place:** changing `purchase_1["price"]` leaves `"price"` where it was in the order. Deleting a key and adding it again puts it at the end.

## List of Dictionaries

The shape used more than others is a list where every element is a dictionary with the same keys. It's how a program holds "many of the same kind of thing".

A purchase record holds all data about one purchase, and a receipt is a list of many purchase records.

Receipt program from earlier lesson rewritten:

````python
purchases = []

while True:
    item = input("Item (or done to finish): ").strip()
    if item.lower() == "done":
        break

    # for now, assume the user enters a valid number
    price = float(input(f"Price of {item}: "))

    # one record holds both pieces of information about the purchase
    purchases.append({"item": item, "price": price})

total = 0
for purchase in purchases:
    print(f"{purchase['item']}: ${purchase['price']:.2f}")
    total += purchase["price"]

print(f"Total: ${total:.2f}")
````

- purchases.append({"item": item, "price": price}) adds one record per purchase.
- for loop hands us one record at a time. `purchase['item']` and `purchase['price']` come from the same record, so they always belong to the same purchase.

> Notice that in {"item": item, "price": price}, the key "item" and the variable item are spelled the same but are two different things. The key is a string inside the dictionary. The variable holds the value that gets stored under that key.

````python
purchases = [
    {"item": "coffee", "price": 3.50},
    {"item": "bagel", "price": 2.25},
    {"item": "juice", "price": 4.00},
    {"item": "muffin", "price": 3.00},
]

# Reach one by position, then by key
print(purchases[1]["price"])  # 2.25

# Remove one the same way you remove from any list
purchases.pop(1)  # the bagel and its price leave together
print(len(purchases))  # 3
````

### Tutor note: why this beats parallel lists

In 1.7 the receipt used two lists, `items` and `prices`, matched by index. That worked, but nothing kept them together.

| | Parallel lists (1.7) | List of dictionaries (1.8) |
|---|---|---|
| One purchase is | `items[1]` and `prices[1]` | `purchases[1]` |
| Reading the price | `prices[1]` (you remember index 1 is the bagel) | `purchases[1]["price"]` (the key says what it is) |
| Removing a purchase | Two `pop`s that must match | One `pop`, the pair leaves together |
| Can get out of step? | Yes, if one list changes without the other | No, the item and price live in one record |

It also fixes the 2D list problem from 1.7: `receipt[1][1]` didn't say what column 1 meant, but `purchases[1]["price"]` does.

## A Dictionary with a List

A dictionary can hold a list of records too. A receipt that also records who made the purchases "nests" the list inside a dictionary:

````python
receipt = {
    "customer": "Ada",
    "purchases": [
        {"item": "coffee", "price": 3.50},
        {"item": "bagel", "price": 2.25},
    ],
}

print(receipt["purchases"][0]["item"])  # coffee
````

- `receipt["purchases"]` is the list of purchase records.
- `[0]` takes the first record in that list, `{"item": "coffee", "price": 3.50}`.
- `["item"]` reads that record's item, `"coffee"`.

### Tutor note: read a chain left to right

Each pair of brackets works on the result of the one before it, so read it one step at a time:

```text
receipt                              -> the whole receipt dictionary
receipt["purchases"]                 -> a list of records
receipt["purchases"][0]              -> {"item": "coffee", "price": 3.50}
receipt["purchases"][0]["item"]      -> "coffee"
```

The type at each step tells you which bracket comes next. A **dictionary** needs a key (`["purchases"]`), and a **list** needs an index (`[0]`). If you use the wrong one, you get a `KeyError` or `TypeError` that tells you which step broke.

Predict, then run. The cashier makes Ada's coffee free by deleting its price. What does the program print?

````python
receipt = {
    "customer": "Ada",
    "purchases": [
        {"item": "coffee", "price": 3.50},
        {"item": "bagel", "price": 2.25},
    ],
}

del receipt["purchases"][0]["price"]

total = 0
for purchase in receipt["purchases"]:
    total += purchase["price"]
print(total)
````

I predicted it to print only 2.25 but:

- `del receipt["purchases"][0]["price"]` removes the `"price"` key from the coffee's record, so the coffee record is now `{'item': 'coffee'}`.
- The loop's first pass reaches the coffee record, and `purchase["price"]` asks it for a key it no longer has. The loop stops with a `KeyError` before `total` is ever printed.

Deleting a key changes the record's shape, and a loop that expects every record to have that key will crash on the record that lacks it. To make the coffee free while keeping its shape, set its price to 0 with `receipt["purchases"][0]["price"] = 0`

### Tutor note: the second intentional error

The cell above also ends with a `KeyError` on purpose. Your prediction of `2.25` makes sense, since only the bagel still has a price. But the loop doesn't skip the record that's missing a key. It crashes on it.

If a record really might be missing a key, `purchase.get("price", 0)` would make the loop treat it as free instead of crashing. Setting the price to `0` is still the better fix, because it keeps every record the same shape.

# 4. Lookup Tables: Finding a Value from a Key

A record's keys are chosen by the programmer in advance. A lookup table's keys are a part of the data. Below is a café menu that maps each item to its price.

````python
menu = {
    "coffee": 3.50,
    "bagel": 2.25,
    "juice": 4.00,
    "muffin": 3.00,
}

print(menu["bagel"])  # 2.25
````

The keys "coffee", "bagel", "juice", and "muffin" are not names the programmer picked to describe a purchase. They are the items the café sells, and when the café adds a new item, the menu gains a new key.

### Tutor note: a quick test for record vs lookup table

Ask: **would the keys be the same no matter what data the program gets?**

- **Yes → record.** Every customer account has `"name"`, `"email"`, and `"is_member"`, whoever the customer is.
- **No → lookup table.** The keys of `loyalty_points` depend on which customers exist. A new customer means a new key.

A second clue is the values. In a record they're usually **different types** (a string, a string, a boolean). In a lookup table they're usually **all the same kind** (all points, all counts, all prices).

The two often work together: a lookup table's values can be records. For example, `customers = {"Ada": {"email": "ada@marcy.org", "is_member": True}}` looks up a customer by name and gets back their record.

## Looking Up a Key Held in a Variable

The brackets can hold any expression. Python evaluates it first, then uses the result as the key.


````python
item = "bagel"
print(menu[item])  # 2.25, the value stored under the key 'bagel'
````

`menu[item]` uses the value of the variable item.
`menu["item"]` uses the literal text "item", which gives a `KeyError` if that key doesn’t exist.
No quotes = variable. Quotes = literal text.

A key held in a variable is what makes a lookup table useful. The program below asks for each item by name and finds its price on the menu, so the user no longer types prices at all:

````python
purchases = []

while True:
    item = input("Item (or done to finish): ").strip().lower()
    if item == "done":
        break

    # Instead of asking for the price, we first check that the item is on the menu...
    if item not in menu:
        # ...print a message if it isn't there...
        print(f"Sorry, {item} is not on the menu")
        continue

    # ...and then use the price from the menu
    purchases.append({"item": item, "price": menu[item]})

total = 0
for purchase in purchases:
    print(f"{purchase['item']}: ${purchase['price']:.2f}")
    total += purchase["price"]

print(f"Total: ${total:.2f}")
````

The program uses both the record and lookup table at once. `menu` is a lookup table that turns an item's name into its price, and each `{"item": item, "price": menu[item]}` is a record of one purchase. `menu[item]` would raise a `KeyError` for an item the café does not sell, such as "pizza", so the `if item not in menu`: check runs first. The check prints a message that tells the user what went wrong, and continue sends the loop back to ask for the next item before `menu[item]` ever runs.

### Tutor note: why `.lower()` matters here

The `input()` line adds `.lower()` to what the user types. Keys are case-sensitive, so without it, typing `Coffee` would fail the `in menu` check even though `"coffee"` is on the menu. Lowercasing the input makes it match the way the keys are stored.

This is the clean-then-validate pattern from 1.5: `.strip()` and `.lower()` clean the input, `if item not in menu` validates it, and only then does `menu[item]` use it.

**Another way to write it:** `price = menu.get(item)` gives `None` for an item that isn't on the menu, so `if price is None:` could replace the `in` check. Both are correct. `in` is the more common choice when the program needs to stop and print a message.

## Challenge: Dynamic Keys

Complete the program below so that it lets the café owner add items to the menu!

````python
menu = {
    "coffee": 3.50,
    "bagel": 2.25,
    "juice": 4.00,
    "muffin": 3.00,
}

while True:
    print("Here is your menu: ", menu)
    new_item = input("Add an item to your menu, or press q to exit. ")

    if new_item == 'q':  # a guard clause
        break

    # for now, assume the owner enters a valid number
    price = float(input(f"Okay, what is the price of {new_item}? "))

    # add the new item and its price to the menu!
    menu[new_item] = price
````

Added `menu[new_item] = price` at the end of the loop, because adding to a dictionary means assigning a value to a new key.

The menu started with four keys, and the program cannot know which items the owner will add. Each new key comes from what the owner types, which is what makes menu a lookup table rather than a record.

### Tutor note: create and update are the same line

`menu[new_item] = price` adds a new item, but if the owner types an item that's already on the menu, like `bagel`, the same line **replaces** its price instead. No error, no duplicate key.

The other inputs are worth knowing too:
- The input isn't cleaned, so `"Bagel "` (capital B, trailing space) becomes a brand new key separate from `"bagel"`. Adding `.strip().lower()` like the ordering program would prevent that.
- The price line assumes a valid number. Typing `abc` raises a `ValueError` from `float()`, the same as in 1.5.

## Iterating Over Keys and Values of a Dictionary

A dictionary can be iterated over which is useful when using a dictionary as a lookup table. A `for` loop over a dictionary visits its **keys**:

````python
menu = {
    "coffee": 3.50,
    "bagel": 2.25,
    "juice": 4.00,
    "muffin": 3.00,
}

for item in menu:
    price = menu[item]
    print(f"{item} costs ${price:.2f}")
````

`item` in the `for` loop is a key in the dictionary and `price` is set to that key's value on each pass with `menu[item]`.

Dictionaries also have three methods that give you exactly the part you want to loop over:

- `menu.keys()` -- all of the keys
- `menu.values()` -- all of the values
- `menu.items()` -- every key and its value, paired up

For example, `.values()` would help us calculate the total without needing to know the item names:

````python
total = 0
for price in menu.values():
    total += price

print(f"One of everything costs ${total:.2f}")  # One of everything costs $12.75
````

This goes through every value in `menu` and adds it to `total`. The item names never come up, because the total only needs prices.

`.items()` is the one used most, because it hands over both halves of each pair, and they can be unpacked right in the `for` line:

````python
for item, price in menu.items():
    print(f"{item} costs ${price:.2f}")
````

Predict, then run. The café wants a list of everything on the menu that costs less than $3.50. What does this program print?

````python
menu = {
    "coffee": 3.50,
    "bagel": 2.25,
    "juice": 4.00,
    "muffin": 3.00,
}

cheap_items = []
for item, price in menu.items():
    if price < 3.50:
        cheap_items.append(item)

print(cheap_items)
````

This will append any items with a price under 3.50 into the new list `cheap_items`:

```python
['bagel', 'muffin']
```

- `.items()` hands the loop one key-value pair at a time, and the `for` line unpacks each pair into `item` and `price`.
- The `if` statement compares each `price` with `3.50`.
- Only the key, `item`, is appended, so `cheap_items` holds item names rather than prices.
- The names appear in the order they were added to the menu, because dictionaries remember the order their keys were added in.

Use `.items()` whenever the loop needs both halves of each pair.

### Tutor note: four ways to loop over a dictionary

| Loop | Each pass gives you | Use it when |
|---|---|---|
| `for item in menu:` | A key | You need the keys (and can look up values with `menu[item]`) |
| `for item in menu.keys():` | A key | Same as above, just more explicit |
| `for price in menu.values():` | A value | You only need values, like adding up a total |
| `for item, price in menu.items():` | A `(key, value)` pair, unpacked | You need both |

**Coffee is not in the list** because `3.50 < 3.50` is `False`. `<` means strictly less than. `<=` would include it.

`.items()` unpacking is the same idea as `for i, item in enumerate(items):` in 1.7. Each pass hands over a pair, and the two names on the left catch its two halves.

**Don't add or delete keys while looping over a dictionary.** Python raises `RuntimeError: dictionary changed size during iteration`. Updating an existing key's value during the loop is fine.

## Challenge: Counting Items

The café wants to know which items sell best. Complete count_items so that it returns a lookup table mapping each item to the number of times it appears in purchases:

````python
purchases = [
    {"item": "coffee", "price": 3.50},
    {"item": "bagel", "price": 2.25},
    {"item": "coffee", "price": 3.50},
    {"item": "juice", "price": 4.00},
    {"item": "coffee", "price": 3.50},
]

def count_items(purchases):
    counts = {}
    for purchase in purchases:
        item = purchase["item"]
        current_count = counts.get(item, 0)
        counts[item] = current_count + 1
    return counts

print(count_items(purchases))  # should print {'coffee': 3, 'bagel': 1, 'juice': 1}
````

Start with an empty dictionary. The first time an item appears it is not a key yet, so think about which way of reading a value can handle a missing key.

- counts starts empty, because the function cannot know in advance which items the purchases contain.
- The first time "coffee" appears, "coffee" is not a key in counts yet. counts["coffee"] would raise a KeyError, but counts.get("coffee", 0) returns the default 0, which is the right count for an item that has not been seen.
- counts[item] = current_count + 1 either adds the key with a count of 1 or replaces the old count with one more.

Each key in counts comes from the data, so counts is a lookup table built from a list of records.

### Tutor note: tracing `count_items`

| Pass | `item` | `counts.get(item, 0)` | `counts` after the pass |
|---|---|---|---|
| Start | | | `{}` |
| 1 | `"coffee"` | `0` | `{'coffee': 1}` |
| 2 | `"bagel"` | `0` | `{'coffee': 1, 'bagel': 1}` |
| 3 | `"coffee"` | `1` | `{'coffee': 2, 'bagel': 1}` |
| 4 | `"juice"` | `0` | `{'coffee': 2, 'bagel': 1, 'juice': 1}` |
| 5 | `"coffee"` | `2` | `{'coffee': 3, 'bagel': 1, 'juice': 1}` |

This **counting pattern** shows up all the time: start with an empty dictionary, and for each thing, get its current count (defaulting to 0) and add 1. It can also be written in one line: `counts[item] = counts.get(item, 0) + 1`.

It's like the running `total = 0` from earlier lessons, except instead of one total, there's one running total per key.

# 5. Dictionaries are Mutable and Held by Reference

When we assign a dictionary to a variable we are storing a reference to the dictionary's location in memory, not the values themselves.

When we assign a variable holding a dictionary to another variable, both variables hold a reference to the same dictionary. A change made through either variable shows up when you look through the other one:

````python
purchase = {"item": "coffee", "price": 3.50}
receipt_line = purchase           # both variables reference the same dictionary
receipt_line["price"] = 3.00      # mutating the referenced dictionary

print(purchase)                   # {'item': 'coffee', 'price': 3.0}
print(receipt_line)               # {'item': 'coffee', 'price': 3.0}
print(purchase is receipt_line)   # True
````

## Changing a Dictionary Inside a Function

Same thing happens when passing a dictionary to a function. The parameter holds a reference to the caller's dictionary, so if the function changes the parameter, it changes the caller's dictionary too.

A function whose job is to change a record can rely on this. `apply_discount` lowers the price of the purchase it is given and returns nothing, because the caller's dictionary already holds the new price:

````python
def apply_discount(purchase, amount):
    purchase["price"] = purchase["price"] - amount

coffee = {"item": "coffee", "price": 3.50}
apply_discount(coffee, 0.50)
print(coffee)  # {'item': 'coffee', 'price': 3.0}
````

### Tutor note: same picture as lists

```text
coffee   ──┐
           ├──► one dictionary: {"item": "coffee", "price": 3.0}
purchase ──┘    (while apply_discount is running)
```

Inside `apply_discount`, the parameter `purchase` and the caller's `coffee` are two names for one dictionary, exactly like the aliases in 1.7. Changing `purchase["price"]` changes the dictionary they share.

The same rule from 1.7 applies too: **mutating** the parameter (`purchase["price"] = ...`) is seen by the caller, but **reassigning** it (`purchase = {}`) only moves the function's own name and leaves `coffee` alone.

`apply_discount` returns `None`, so `coffee = apply_discount(coffee, 0.50)` would set `coffee` to `None`. It's the same trap as `items = items.append(x)`.

## Copying a Dictionary Before Changing It

To leave the caller's dictionary unchanged, have the function copy the dictionary first. The `dict()` function copies the key-value pairs of one dictionary into a new dictionary, and the function can then change the copy as much as it likes:

````python
def discounted_copy(purchase, amount):
    new_purchase = dict(purchase)  # copy the key-value pairs of purchase into a new dictionary
    new_purchase["price"] = new_purchase["price"] - amount
    return new_purchase

coffee = {"item": "coffee", "price": 3.50}
sale_coffee = discounted_copy(coffee, 0.50)

print(coffee)                 # {'item': 'coffee', 'price': 3.5}
print(sale_coffee)            # {'item': 'coffee', 'price': 3.0}
print(coffee is sale_coffee)  # False
````

`purchase.copy()` and `{**purchase}` do the same thing as `dict(purchase)`.

### Tutor note: change it or copy it

| | `apply_discount` | `discounted_copy` |
|---|---|---|
| Changes the caller's dictionary? | Yes | No |
| Returns | `None` | The new dictionary |
| Caller writes | `apply_discount(coffee, 0.50)` | `sale_coffee = discounted_copy(coffee, 0.50)` |
| Use when | The caller wants the change | The caller needs the original kept |

The three copy styles match the list versions from 1.7:

| Lists | Dictionaries |
|---|---|
| `list(items)` | `dict(purchase)` |
| `items[:]` or `items.copy()` | `purchase.copy()` |
| `[*items]` | `{**purchase}` |

`*` unpacks a list's elements, and `**` unpacks a dictionary's key-value pairs. `{**purchase, "price": 3.00}` copies the record and overrides the price in one step.

# 6. Advanced Topics

## Which Values Can Be Keys

Keys can also be numbers, which comes in handy when you want to look something up by number, like a Roman numeral:

````python
roman_numerals = {
    1: "I",
    5: "V",
    10: "X",
}
````

A key has to be a value that cannot change: a string, a number, or a tuple. A dictionary finds a value by looking up its key, and a key that changed after it went in could no longer be found. A list can change, so a list cannot be a key. 

A tuple can be a key because a tuple cannot change. A tuple key lets a lookup table use two pieces of information at once, such as an item and its size:

````python
sized_menu = {
    ("coffee", "small"): 3.00,
    ("coffee", "large"): 3.50,
    ("juice", "small"): 3.25,
    ("juice", "large"): 4.00,
}

print(sized_menu[("coffee", "large")])  # 3.5
````

### Tutor note: what "cannot change" really means for keys

Python finds a key by turning it into a number called a **hash** and using that number to find the key's spot, which is why it's fast even with thousands of keys. If a key could change, its hash would change, and Python would look in the wrong spot. That's why keys must be **hashable**.

| Key | Allowed? | Why |
|---|---|---|
| `"coffee"` | Yes | Strings are immutable |
| `10`, `3.5` | Yes | Numbers are immutable |
| `True` | Yes | Booleans are immutable |
| `("coffee", "large")` | Yes | A tuple of immutable values |
| `["coffee", "large"]` | No | Lists are mutable: `TypeError` (`unhashable type: 'list'`) |
| `("coffee", ["oat milk"])` | No | A tuple holding a list is still unhashable |

**Only keys have this rule.** Values can be anything, including lists and dictionaries, which is how `receipt["purchases"]` holds a list.

**Watch out:** `1` and `1.0` are equal, so they count as the **same key**. `{1: "I", 1.0: "one"}` ends up with just one key, holding `"one"`.

## A Copy Shares Its Nested Values

`dict()`, `.copy()`, and `{**my_dict}` all make a shallow copy: they copy only the outer dictionary. If a value inside it is a list or another dictionary, the copy and the original share that one inner list or dictionary, so changing it through the copy changes the original too.

Suppose a customer orders a second coffee like their first one, but with an extra shot. Copying the first coffee and adding to the copy's `add_ons` changes the first coffee as well:

````python
first_coffee = {"item": "coffee", "price": 3.50, "add_ons": ["oat milk"]}
second_coffee = dict(first_coffee)
second_coffee["add_ons"].append("extra shot")

print(first_coffee["add_ons"])  # ['oat milk', 'extra shot']
print(first_coffee["add_ons"] is second_coffee["add_ons"])  # True
````

To change a nested list without touching the original, copy that list as well before you change it: `second_coffee["add_ons"] = list(first_coffee["add_ons"])`.

### Tutor note: draw the shared list

```text
first_coffee  ──► { "item": "coffee", "price": 3.5, "add_ons": ─┐ }
                                                                ├──► ["oat milk", "extra shot"]
second_coffee ──► { "item": "coffee", "price": 3.5, "add_ons": ─┘ }
```

`dict(first_coffee)` made a **new outer dictionary**, so the two coffees are separate dictionaries (`first_coffee is second_coffee` is `False`). But the `"add_ons"` value was copied as a **reference**, so both point at the same one list. `.append()` mutates that list, and both coffees see it.

| Change on `second_coffee` | Affects `first_coffee`? |
|---|---|
| `second_coffee["price"] = 4.00` | No: replaces a value in the new outer dictionary |
| `second_coffee["add_ons"] = ["almond milk"]` | No: points the key at a new list |
| `second_coffee["add_ons"].append("extra shot")` | **Yes**: mutates the shared list |

It's the same shallow copy idea as copying a 2D list in 1.7, one level deeper.

# 7. Summary

- A **dictionary** stores values under **keys** instead of positions, so a program looks a value up by what it means rather than where it sits. Keys are unique, and values can be anything.
- Dictionaries do two jobs:
    - A **record** keeps the pieces of information about **one thing** together, like one purchase's item and price. The programmer picks its keys in advance so every record has the same shape.
    - A **lookup table** finds a value from a key, like the price of any item on a menu. Its keys come from the data, so it grows as the program runs.
- Read with **brackets** (`my_dict["key"]`, `KeyError` if missing), **`.get()`** (`None` or a default if missing), or check first with **`in`**, which only checks keys.
- Dictionaries are **mutable**. `my_dict["key"] = value` both creates and updates. `del` and `.pop()` delete, and `.pop()` also returns the removed value.
- A **list of dictionaries** is the most common shape: one record per thing, all with the same keys. It replaces parallel lists, because an item and its price can't drift apart.
- Nested data is read **left to right**: `receipt["purchases"][0]["item"]`. Use keys for dictionaries and indexes for lists.
- Loop over a dictionary's keys with `for key in my_dict`, or use `.keys()`, `.values()`, and `.items()`. Dictionaries remember the order keys were added.
- Dictionaries are **held by reference**, exactly like lists. A function that mutates a dictionary parameter changes the caller's dictionary, so copy it first with `dict()`, `.copy()`, or `{**my_dict}` when the original must stay the same. These are **shallow** copies, so nested lists and dictionaries are still shared.
- Keys must be **hashable** (strings, numbers, booleans, tuples of those). Lists can't be keys.

# 8. Quick Reference

**Reading a dictionary** (`purchase = {"item": "coffee", "price": 3.50}`)

| Code | Result | Notes |
|---|---|---|
| `purchase["item"]` | `'coffee'` | `KeyError` if the key is missing |
| `purchase.get("discount")` | `None` | Never raises an error |
| `purchase.get("discount", 0)` | `0` | Your own default |
| `"price" in purchase` | `True` | Checks **keys** only |
| `"coffee" in purchase` | `False` | `"coffee"` is a value |
| `"coffee" in purchase.values()` | `True` | Checks values |
| `len(purchase)` | `2` | Number of key-value pairs |

**Changing a dictionary** (all of these mutate the original)

| Code | What it does | Returns |
|---|---|---|
| `purchase["discount"] = 0.50` | Adds the key if new, replaces the value if not | — |
| `del purchase["discount"]` | Removes the pair | — |
| `purchase.pop("price")` | Removes the pair | The removed value |
| `purchase.pop("price", None)` | Removes the pair if it's there | The value, or the default |

**Looping**

| Loop | Gives you |
|---|---|
| `for key in my_dict:` | Each key |
| `for value in my_dict.values():` | Each value |
| `for key, value in my_dict.items():` | Each key and value |

**Common patterns**

```python
# list of records
purchases = [{"item": "coffee", "price": 3.50}, {"item": "bagel", "price": 2.25}]
purchases[1]["price"]                  # 2.25

# nested data
receipt = {"customer": "Ada", "purchases": purchases}
receipt["purchases"][0]["item"]        # 'coffee'

# look up with a variable (no quotes!)
item = "bagel"
menu[item]                             # 2.25

# counting
counts = {}
for purchase in purchases:
    counts[purchase["item"]] = counts.get(purchase["item"], 0) + 1
```

**Copies**

```python
alias = purchase                 # same dictionary, two names
copy1 = dict(purchase)           # new dictionary
copy2 = purchase.copy()          # new dictionary
copy3 = {**purchase}             # new dictionary
sale = {**purchase, "price": 3}  # copy + override one key
```

- All copies are **shallow**: nested lists/dictionaries are shared. Copy the nested list too before mutating it.

**Common errors**

| Error | Usually means |
|---|---|
| `KeyError: 'x'` | Key `'x'` isn't in the dictionary (typo, never added, or deleted) |
| `TypeError: ... (unhashable type: 'list')` | Tried to use a list as a key. Python 3.14 says `cannot use 'list' as a dict key`; older versions just say `unhashable type: 'list'` |
| `RuntimeError: dictionary changed size during iteration` | Added or deleted keys while looping over the dictionary |

**Record vs lookup table**

| | Record | Lookup table |
|---|---|---|
| Describes | One thing | Many things |
| Keys | Chosen in advance (`"item"`, `"price"`) | Come from the data (`"coffee"`, `"bagel"`) |
| Grows at runtime? | No, same shape every time | Yes |
| Example | `{"name": "Ada", "is_member": True}` | `{"Ada": 50, "Zo": 25}` |

**Lists vs dictionaries**

| | List | Dictionary |
|---|---|---|
| Written as | `['coffee', 3.50]` | `{"item": "coffee", "price": 3.50}` |
| Look up by | Index (position) | Key (name) |
| Missing lookup | `IndexError` | `KeyError` |
| `in` checks | Elements | Keys |
| Mutable? | Yes | Yes |
| Copy with | `list(x)`, `x[:]`, `[*x]` | `dict(x)`, `x.copy()`, `{**x}` |
