## 1. Data Types

All values have a **data type** that determines what you can do with that value.

In Python, there are five basic **data types** to start with: `str`, `int`, `float`, `bool`, and `None`.

| Data Type | Syntax Example | What it Represent | Common Use Cases|
|---|---|---|---|
| **`str`** | `"morning"` or `'@abc123'` | Text (any sequence of characters) | Usernames, chat messages, titles, product descriptions, search queries etc. |
| **`int`** | `100` or `-15` | Whole numbers | Scores in a game, health, count of lines, a tasks position on a list. | 
| **`float`** | `10.5` or `0.99` | Numbers with a decimal point. | Prices, averages, percentages, timers. |
| **`bool`** | `Truth` or `False` | Truth value | Login status, feature toggles (dark mode on/off), win/lose conditions. |
| **`None`** | `None` | Intentional absence of a value | Representing "no result" or "not yet set" |

**`lists`**, **`dictionaries`**, and **`functions`** are the three types that hold other data or hold code.

| Data Type | Syntax Example | What it Represent | Common Use Cases|
|---|---|---|---|
| **`list`** | `["apple",<br>"banana",<br>"cherry"]` | An ordered list of similar values. | Storing lists (shopping cart items, game high scores, chat messages) |
| **`dict`** | `{"name": "Alex", <br>"age": 25}` | A collection of data relating to one "thing" | Storing user profile data,  storing data about product (name, description, price, etc...) |
| **`function`** | `def greet():<br> print("Hi!")` | A reusable block of code | Performing action, handling events, calculations> |

# 2. Operators

An **operator** is a symbol like `+`, `*`, `>`, or `==` that takes values and produces a new one. The functionality of the symbol depends on the **data type** of the values you give it. 

Theres more than one way to store the same information. A light switch could be the text "on" or "off" or the Boolean True or False. This is picked based on what you need to do with it later, since each type works with different operators. Booleans work directly with `if`, `and`, `or`, and `not`, the text version would need comparing every time `(if switch == "on")`.

```python
5 + 2               # 7: + adds numbers
"hello" + " world"  # "hello world": + joins strings
5 * 2               # 10: * multiplies numbers
"hello" * "world"   # TypeError: you can't multiply text by text
```
`+` works on two numbers and on two strings but they do different jobs. `*` works on two numbers but with two strings it means nothing and throws a `TypeError`. That error comes up when an operator is given types it cant handle.

````python
print(10 + 5 * 2)            # 20: multiplication happens before addition
print("B" > "A")             # True: Letters coming after are greater, lowercase is always after capital. "B" comes after "A" in the alphabet
print(2 > 1 and "B" > "A")   # True: both sides are True
print(8 / 2)                 # 4.0: / always gives a float
print("3" * 5)               # 33333: a string times a number repeats the string
print(True + True + False)   # 2: True counts as 1, False counts as 0
print([1, 2] * 2)            # [1, 2, 1, 2]: a list times a number repeats the list
print(0 or "Python")         # Python: or skips 0 (counts as false) and gives the other value
print(0.1 + 0.2 == 0.3)      # False: decimals have tiny rounding errors
print([1, 2] == [1, 2])      # True: == checks for the same contents
print([1, 2] is [1, 2])      # False: is checks for the same object, and these are two separate lists
````

# Resolution Order of Operations

**Operator Precedence** is the order in which operators are evaluated. Understanding this is crucial.

````python
fahrenheit = 212
celsius = fahrenheit - 32 * 5 / 9
print(celsius)  # 194.22222222222223
````

We can use this example again. The output should be 100°C.

The **Order of Precedence** is:

1. Anything inside parentheses `()` is evaluated first.
2. Then arithmetic operators, `**` is evaluated before `*`, `/`, `//`, `%`, which are evaluated before `+` and `-`. Operators at the same level are evaluated left to right.
3. Followed by comparison operators (`<`, `<=`, `>`, `>=`, `==`, `!=`).
4. Then logical operators. `not` is evaluated first, then `and`, then `or`.
5. Then conditional expressions (`x if condition else y`) and assignment happen last (`=`).
