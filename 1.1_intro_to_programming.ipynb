# 2. What is a program?

A program is a text file with instructions that a computer executes to accomplish some task. It will be made up of comments, expressions, and statements:

### **Comments** (`#`) 
are ignored and help to explain code or annotate your code.

```python
# this is a comment
```
### **Expressions** 
are a piece of code that evaluates to a single value, results of operations (`5 + 5`), a function call (`len("Hi")`), and standalone values like string literal `"Hello WOrld"` are also considered expression because they evaluate themselves.

````python
5 + 5 # Evaluates to 10
len("hi") # Evaluates to 2
"hello" # Evaluates to "hello"
5 # Evaluates to 5
````

Expressions on their own do nothing as shown above. They become useful when they are used within statements. Like a `print()` statement

### **Statements**: 
are instructions that perform an action. They change the program in some way, often using expressions. 

Examples are: Variables assignments alter the program's **state** which is the data that is stored by a program at a point in time. It gets stored in the program's memory

`if/else` statements change the control flow of the program.

````python
# assigning a variable stores the value in the programs memory to be used later
student = "xavier"
mood = None 

# if statements change the control flow of the program(the order the commands are being executed) in based off a conditional
if student != "xavier":
    mood = "sad"
else:
    mood = "happy"

print(mood)
````

# 3. **Running a file with the Python Interpreter**

### **Python Interpreter**
The **Python Interpreter** is a piece of software that runs the code in a file. It reads a `.py` file and executes it, one statement at a time.

The interpreter can be activated in the Terminal using the `python3 [filename]` command:

```python
# main.py is the name of the .py file we want to run
python3 main.py
```

# 4. **Printing to the Terminal with `print()`**
### **`print()`**
Is a built -in function that prints text to the Terminal.

````python
print('hello world')
````

**`print()`** is primarily used to verify the output of a program for debugging purposes.

Example: the code below converts 212 Fahrenheit to Celsius which should produce the result 100C. 

Do you expect it to work?

````python
fahrenheit = 212
celsius = fahrenheit - 32 * 5 / 9
````

The code above does **NOT** show any outcome in the terminal so we don't know whether its correct or not. Thats when `print()` comes into play.

With **`print()`** we can verify whether or not the calculation was performed properly.

````python
fahrenheit = 212
celsius = fahrenheit - 32 * 5 / 9
print(celsius)
````

This did not give us the answer we wanted. PEMDAS applies here in python as well. The fix is parentheses.

````python
fahrenheit = 212
celsius = (fahrenheit - 32) * 5 / 9
print(celsius)
````

Without **`print()`** we wouldn't have known there was a bug with the calculation and thats exactly what its for.
