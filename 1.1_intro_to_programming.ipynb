# 1. Key Terms

| Term | Meaning |
|---|---|
| **Program** | A  program is a text file containing instructions that a computer executes to accomplish a task. |
| **Comments** | Used to document and explain the code. They are ignored when a program is executed. A comment line begins with the `#` symbol and can be placed on its own line or following valid code. |
| **Expressions** | are any piece of code that evaluates to a single value. They are often the result of using operators (`+,*,>, ==` etc...) or invoking functions (`len("hi")`). |
| **Statements** | These change the program.<br>• Creating a variable increases te memory used by the program.<br>• `if` statements change the "control flow" of a program by skipping lines of code |
| **State** | refers to the data stored by a program ata  point in time. |
| **Python Interpreter** | Program that reads a `.py` file and executes it, one statement at a time. Start with the `python3` command. |
| **`print()`** | The `print()` function is a built-in function that displays text to the Terminal output. It is used to view the result of code operations, primarily for easily debugging purposes and simple testing. |
| **`f-string`** | A string with an `f` before the opening quotation, Anything written inside `{}` is replaced by its value. `x = 10 print(f"x is {x}") # prints x is 10`|
