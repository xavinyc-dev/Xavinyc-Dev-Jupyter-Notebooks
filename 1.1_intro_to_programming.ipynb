# 2. What is a program?

A program is a text file with instructions that a computer executes to accomplish some task. It will be made up of comments, expressions, and statements:

### **Comments** (`#`)
Comments are ignored by Python. They help to explain or annotate your code.

```python
# this is a comment
```
### **Expressions**
Expressions are pieces of code that evaluate to a single value. Operations (`5 + 5`) and function calls (`len("Hi")`) are expressions. Standalone values like the string literal `"Hello World"` are also considered expressions, because they evaluate to themselves.

````python
15 - 5 # Evaluates to 10
len("hello") # Evaluates to 5
"hello world" # Evaluates to "hello world"
5 # Evaluates to 5
````

Expressions on their own do nothing in a `.py` file: Python works out the value and then throws it away. They become useful when they are used within statements, or passed to a function like `print()`.

> **Note:** a notebook cell automatically displays the value of its **last line**, which is why `5` appears above. Running the same lines with `python3` would show nothing.

### **Statements**
Statements are instructions that perform an action. They change the program in some way, often using expressions.

Examples: variable assignments alter the program's **state**, which is the data stored by a program at a point in time. It gets stored in the program's memory.

`if/else` statements change the control flow of the program.

````python
# assigning a variable stores the value in the program's memory to be used later
last_years_tips = 201
mood = None 

# if statements change the control flow of the program (the order the commands are executed) based on a condition
if last_years_tips <= 200:
    mood = "sad"
else:
    mood = "happy"

print(mood)
````

# 3. **Running a file with the Python Interpreter**

### **Python Interpreter**
The **Python Interpreter** is a piece of software that runs the code in a file. It reads a `.py` file and executes it, one statement at a time.

The interpreter can be activated in the Terminal using the `python3 [filename]` command:

```bash
# main.py is the name of the .py file we want to run
python3 main.py
```

# 4. **Printing to the Terminal with `print()`**
### **`print()`**
`print()` is a built-in function that prints text to the Terminal.

````python
print('welcome to marcy')
````

**`print()`** is primarily used to verify the output of a program for debugging purposes.

Example: the code below converts 212° Fahrenheit to Celsius, which should produce the result 100°C.

Do you expect it to work?

````python
fahrenheit = 212
celsius = fahrenheit - 32 * 5 / 9
````

I think the code runs, but it does not output anything to the Terminal, so I can't tell yet whether the answer is right.

The code above does **NOT** show any outcome in the Terminal, so we don't know whether it's correct or not. That's when `print()` comes into play.

With **`print()`** we can verify whether or not the calculation was performed properly.

````python
fahrenheit = 212
celsius = fahrenheit - 32 * 5 / 9
print(celsius)
````

This did not give us the answer we wanted. PEMDAS applies here in Python as well: multiplication and division happen before subtraction, so Python calculated `32 * 5 / 9` first and then subtracted that from `212`. The fix is parentheses.

````python
fahrenheit = 212
celsius = (fahrenheit - 32) * 5 / 9
print(celsius)
````

Without **`print()`** we wouldn't have known there was a bug with the calculation, and that's exactly what it's for.

# 5. **Printing Values Inside Text with `f-strings`**

To print a value with text around it, you must put an **`f`** in front of the opening quote of a **string**. Python will fill in any expression you write inside curly braces **`{}`** that are inside that string.

````python
fahrenheit = 212
celsius = (fahrenheit - 32) * 5 / 9

print(f"{fahrenheit}°F is {celsius}°C")
````

The string above is called an **`f-string`**. Without the `f`, it would just be braces and text, with no values.

# 6. **Control Flow**

Computers can only execute one statement at a time. **Control Flow** refers to the order in which the program's statements are executed.

The default is top to bottom.

````python
print("1")
print("2")
print("3")
````

There are tools, such as `if` statements, functions, and loops, that let us alter the program's control flow. For example, `if` statements can cause certain statements to be skipped.

````python
day_off = True
mood = None

# if statements change the control flow of a program
if day_off != True: # if does not equal
    mood = "sad" # this is skipped 
else:
    mood = "happy" # this line of code is executed next instead
print(mood)
````

A `for` loop can cause one or more statements to be executed more than once.

````python
# the indented line runs once for each number: 1, 2, 3
for number in [1, 2, 3]:
    print(f"lap {number}")

print("done")  # runs once, after the loop finishes
````

# 7. **Code Style and Readability**

**Readability** is how easy it is for another engineer to read and understand your code.

Python names use `snake_case`: lowercase words joined by underscores.

````python
# hard to read: the names say nothing about what the values are
x = 212
y = (x - 32) * 5 / 9

# easy to read: snake_case names that describe the values
temp_fahrenheit = 212
temp_celsius = (temp_fahrenheit - 32) * 5 / 9

print(temp_celsius)
````

Both versions calculate the same thing. The second one can be understood without any comments, because the names explain it.
