## `if/elif/else` statements

To define the possible decisions the program makes we can use `if`, `elif`, and `else` statements.

These statements set the conditions for the program to follow. (`if this condition is met then do that, else if not met do this, else if none is met do this.`)

For example this function uses these if statements to print a message:

```python
def is_it_hot(temp):
    if temp > 100:
        print("So Hot!")
    elif temp > 90:
        print("Yes!")
    elif temp > 75:
        print("Eh")
    else:
        print("Nah")

is_it_hot(80)
# Output: Eh

is_it_hot(95)
# Output: Yes!

is_it_hot(105)
# Output: So Hot!

is_it_hot(60)
# Output: Nah
```

-  The condition after `if` or `elif` is usually a comparison like `temp > 100`. The code block under it, which we can visually see with an indent, only runs if this condition is met.
- `elif` is short for else if and only runs if the previous conditions wre `False` and its own condition is `True`
- There can be multiple elifs or none at all.
- `else` statements do not require a condition and will run only when none of the other statements were met.

## Order Matters

Python checks the conditions from top to bottom as control flow does and runs the code block under the first condition that is True, which means if we have 4 `if`/`elif` statements and only the last is true then it will only run the lat last statement. As soon as one block runs, Python skip every remaining `elif` and `else` without checking them. If we have a broad condition like `temp > 75` comes first, it catches 95 and 105 and then narrower conditions below it never get a turn.

```python
def is_it_hot(temp):
    if temp > 75:
        print("Eh")
    elif temp > 90:
        print("Yes!")
    elif temp > 100:
        print("So Hot!")
    else:
        print("Nah")

is_it_hot(80)
# Output: eh

is_it_hot(95)
# Output: eh

is_it_hot(105)
# Output: eh

is_it_hot(60)
# Output: nah
```

The first function calls input a number that meets the condition of the first if statement `temp > 75` so it runs that line and never runs the others. Where as 60 was compared with every if statement and came back false for each so it resorted to the else statement.
