# 1. Key Terms

| **Terms** | **Definitions** |
|---|---|
| **Strings** | Sequences of characters enclosed in quotes (single or double). Strings are immutable, meaning all operations are "read-only". |
| **`in` Operator** | Checks whether one string contains another. |
| **Bracket Notation** | Returns individual characters within a string: `str[index]` |
| **`len()`** | The built-in `len()` function returns the number of characters in a string (and you can use this method on lists and dictionaries too!). |
| **Method** | A function that is attached to a value. Methods are invoked using **dot notation**: `value.method()`. |
| **String Methods** | String methods like `startswith`, `endswith`, `find`, `upper`, `replace`, `strip`, and `split` allow you to search, analyze, and manipulate string data. |
| **`.index()` (lists)** | Lists don't have `.find()`. `list.index(x)` returns the index of the first match, and crashes with a `ValueError` when there's no match instead of returning `-1`. |
| **`print()`** | Takes any number of values. It puts `sep` between them, a space unless you say otherwise, and `end` after them, a new line unless you say otherwise. |
| **f-strings** | Interpolate the value of any expression written inside `{}` into a string. A **format specifier** after a colon, like `{price:.2f}`, controls how the value is displayed. |
| **`input()`** | The `input()` function gets input from the user. It always returns a string, so casting it to another type is your job. |
| **Type Casting** | Type casting with `str()`, `int()`, `float()`, `bool()` turns a value of one type into another. |
| **Hard-coded** | A program is hard-coded if the program code must be modified in order to produce a new result. |

# 3. Input

A program is considered **hard-coded** if the program must be modified in order to produce a new result.

The built-in `input()` function lets the user running the program provide the data instead.
The `input()` function will always return the input as a `string`.

When writing the data ourselves we know exactly what each value held. A user can type anything, extra spaces, unexpected capitals, nothing at all, or an email address with no `@`.

A program that uses those values exactly will print strange outputs or crash. We can control what the program does with that value for this reason. A program typically does these things with user input:

1. cleaning
2. validating
3. parsing

## Getting Input with `input()`

`input("prompt: ")` prints out the prompt to the user, then waits for the user to type a line and press `Enter`, and returns what they typed back to the program.


````python
name = input("Hello there! What is your name? ")
print(f"Hi {name}. My name is Bob")
````

Space at the end of the `input()` prompt is so the user's input and the prompt are separated when the user types.

## Cleaning Input

**Cleaning** an input is fixing up what the user types in order for the program to use it the way we want it to.

People don't type the way our program expects them to. They can write something in all caps, accidentally press space, or type something like `"Yes"` instead of `"yes"`. If `answer == "yes"` is our comparison then `"Yes"` would return as `False` as Python compares strings character by character. So the first character being capitalized in the user's input will not match the comparison and without cleanup the program would not run that block.

Strings come with tools for cleaning up text called **methods**.

**Methods** are functions that are attached to a value. Methods are invoked or called using **dot notation**: `value.method()`:

**`.strip()`** returns the string without the whitespace on either end. When you put something in the `()`, `.strip()` removes those characters from the ends instead of whitespace

**`.lower()`** returns the string with every character lowercase

**`.upper()`** returns the string with every character capitalized

```py
answer = '  YES '

# str.strip() returns `str` without the whitespace at either end
print(answer.strip())
# Output: YES

# str.lower() returns `str` with every character in lowercase
print('YES'.lower())
# Output: yes

# str.upper() returns `str` with every character capitalized
print('yes'.upper())
# Output: YES

# Other data types have methods too, such as float.is_integer()
print(5.0.is_integer()) # True
print(5.1.is_integer()) # False
```

Strings are **immutable** meaning they can not be changed. So string methods never change the string they are called on. Methods like `strip()` and `lower()` first copy the string they are operating on and return a new string instead, leaving the original string untouched. So after `answer.strip()` runs, `answer` still holds `'  YES '`.

````python
answer = '  YES '
answer.strip()
answer.lower()
print(answer)
# Output:   YES   (the spaces and the capitals are still there)
````

````python
answer = '  YES '
cleaned = answer.strip().lower()
print(cleaned)
# Output: yes
````

This works because `strip()` returns a new string, and `.lower()` is called on that new string.

**Method Chaining** lets us use multiple methods on the same string. This lets us clean input on the same line as the `input()`:

````python
answer = input("Do you want to continue? (yes/no) ").strip().lower()

if answer == "yes":
    print("Let's keep going!")
else:
    print("Okay, see you later.")
````

For this instance whether the user types `yes`, `YES`, or `Yes`, `answer` holds `"yes"`.

If the user typed `  YES  ` into the input, `answer` would equal `"  YES  ".strip().lower()`. It will then clean that string in order of the methods. `.strip()` returns a new string taking away the space from both sides. `answer = "YES".lower()`. Then the `.lower()` method will use that string and return it with all lowercased characters. `answer = "yes"`.

Sometimes cleaning means removing characters from the middle of the string instead of the ends.

`replace(old, new)` replaces every appearance of `old` with `new`, so replacing a character with an empty string like this removes it.

````python
phone = '555-123-4567'

print(phone.replace('-', ''))
# Output: 5551234567

print('apples'.replace('p', 'g'))
# Output: aggles

print('apples'.replace('p', 'g', 1))
# Output: agples
````

`replace` replaces every match unless you tell it how many to replace with a third argument.

### Type Casting

Cleaning input also means giving it the data type the program needs.

````python
age = input("How old are you? ")
print(age * 2)
````

**`input()` always returns a string no matter what the user typed.** The user typed `20`, but `age` holds `"20"`, and a string multiplied by `2` is repeated.

Every time you take input from a user and want a number, casting it is your job. Nothing does it for you.

Every value that comes out of `input()` is a string, even when the user typed a number. If `age` holds the string `"20"` both arithmetic and comparison operators would crash with a `TypeError` as Python does not do those between a string and an integer.

**Type casting** is when we convert one data type to another data type. We can cast a value by using a casting function like `str()`, `int()`, and `float()`.

````python
# Everything can easily be turned into a string
print(str(1))          # -> "1"
print(str(True))       # -> "True"

# Not everything becomes a number in a nice way
print(int(False))      # -> 0
print(int("42"))       # -> 42
print(float("0.50"))   # -> 0.5
print(int("hello"))    # -> crashes with ValueError: invalid literal for int() with base 10: 'hello'
````

In the above example we can only cast numeric characters or digits from a string to an `int` or `float`.

We can cast directly on the `input()` like we did with string methods:

````python
age = int(input("How old are you? ").strip())
print(f"Next year you will be {age + 1}")
````

It will take the input from me as a string, then strip it and send back a new string, followed by casting it with `int()` to turn it into an integer for the program to use.

If the user types `"twenty"` instead of `20`, `int()` crashes with the same `ValueError`. 

The fourth type casting function is `bool()`, the one used with truthy and falsy values in chapter 1.4.

## Validating Input

**Validating** an input means checking the input is acceptable for the program to actually use it. Clean input can still be the wrong input. If the user inputs nothing at all or typed an email with no `@` while the program looks for the character specifically then the program can't use the input as we intended.

Methods are called on a value: `value.method()`
A function returns a value.
The `input()` function returns whatever is typed as a string.

`isnumeric()` checks for standard numbers, superscripts/subscripts like to the power, fractions, roman numerals and foreign numeral characters.

### The `len()` builtin

The `len()` function returns the number of characters in a string including spaces. Since `len()` is a function we input the string inside the parentheses as the argument.

````python
message = 'Hello there!'
message_space = '     Hi! there.     '
print(len(message))
# Output: 12

print(len(message_space))
# Output: 20

print(len(''))
# Output: 0
````

A string with no characters at all is an empty string and its length is `0`. Pressing Enter without typing anything in `input()` will return an empty string. With `strip()` even if they only press space then it'll return empty as strip will remove that space.  

````python
name = input("What's your name? ").strip()

if len(name) == 0:
    print("You didn't type a name.")
else:
    print(f"Hello, {name}!")
````

Above I input an empty string. Without the check or `guard clause` then the program would have printed `Hello, !`. Checking the length first is a form of validation, and prints to the user what went wrong so they input what's needed.

````python
print("book" in "bookshelf")

print("BOOK" in "bookshelf")
````

This operation above is case sensitive. If the user input BOOK or any form of capitalization then it would come back false. That's why we clean the input with `.lower()` before we validate it's inside the string.

### `startswith()` and `endswith()`

If where the text appears matters then `.startswith()` and `.endswith()` methods check only the start and end of the string. You input what substring or characters you are looking for in the parentheses.

````python
website = 'https://marcylabschool.org'

print(website.startswith('https://'))
# Output: True

print(website.endswith('.com'))
# Output: False
````

### `isdigit()`

The `.isdigit()` method checks whether a string is made entirely of digits. `isdigit()` is the check to run before casting a string into an integer as non digit characters will crash the program.

>**Note** that `.` is a non digit character so floats in strings will return False in the isdigit check.

````python
print('42'.isdigit())
# Output: True

print('4.2'.isdigit())
# Output: False
````

Validating a string before casting lets us give the user a message instead of crashing. `.isdigit()` returns `False` for any non digit character including `' '` a space and `-`, `.`. We can use `.strip()` before validating to resolve the space issue.

````python
age = input("How old are you? ").strip()

if not age.isdigit():
    print(f"Sorry, {age} is not a number")
else:
    age = int(age)
    print(f"Next year you will be {age + 1}")
````

Above has a **guard clause** that validates if the input the user has given is a number using `.isdigit()` and if not it prints a message back to the user rather than crashing.

## Parsing Input

**Parsing** input means taking apart what the user gives us to get the pieces the program needs. For example, the first letter of a name or a part of an email address before the `@`. Taking pieces out of a string starts with knowing where each character is.

### Bracket Notation

Each character in a string including spaces has an **index**, which is a numbered position starting at 0.

We can get a single character from a string using **Bracket Notation** and the index: `string[index]`

````python
print("abc"[0])    # Output: a
print("abc"[1])    # Output: b
print("abc"[2])    # Output: c
````

Can use it directly on a variable as well:

````python
message = 'Hello there!'

print(message[0])   # Output: H
print(message[4])   # Output: o
print(message[5])   # Output:   (a space)
print(message[11])  # Output: !
````

Python also lets you count backwards from the end with a negative index starting at `-1` as the last character.

````python
print(message[-1])  # Output: !
print(message[-2])  # Output: e

print(message[len(message) - 1])
# Output: !  (the same character as message[-1])
````

### Slicing

Indexing takes out one character whereas **Slicing** with `[start:end]` takes out a piece of the string, from `start` up to but not including `end`:

````python
games = 'cyberpunk, minecraft, stardew valley' 

print(games[0:9])
# Output: cyberpunk
````

Either number in a slice can be left out, it will default from the first character if `start` is left blank and the last character if `end` is left blank. A negative number counts from the end of the string.

````python
print(games[:9])
# Output: cyberpunk

print(games[11:])
# Output: minecraft, stardew valley

print(games[-6:])
# Output: valley
````

### Slicing with `find()`

A slice needs index numbers, but with input the user decides where the part of the string we need is. 

The methods `find` and `rfind` return the index of a character or a longer piece of text that we are looking for.

- `find` returns the index of the first occurrence
- `rfind` returns the index of the last occurrence
- Both methods count the index from the left, when there's no match it returns `-1`.

````python
game = 'stardew valley'

print(game.find('a'))
# Output: 2

print(game.rfind('a'))
# Output: 9

print(game.find('y'))
# Output: 13
````

### `.index()` for Lists

Lists don't have a `.find()` method. To get the index of something in a **list**, we use `.index()` instead. It returns the index of the first match, counting from `0` just like string indexes:

````python
haystack = ["hay", "junk", "hay", "needle", "moreJunk"]

print(haystack.index("needle"))
# Output: 3

# haystack.find("needle")   # crashes with AttributeError: lists have no .find()
````

The big difference is what happens when there's no match:

| | **When it's found** | **When it's not found** |
|---|---|---|
| `string.find(x)` | Returns the index | Returns `-1` |
| `list.index(x)` | Returns the index | Crashes with a `ValueError` |

So if the item might not be in the list, check with `in` first, the same way we validate the `@` with `in` before using `find("@")` below:

```py
if "needle" in haystack:
    print(haystack.index("needle"))
```

>**Note:** Strings have an `.index()` method too. It works like `.find()` but crashes with a `ValueError` instead of returning `-1`. Lists are covered fully in chapter 1.7.

`find()` and slice take out a piece of the input whose position we could not know in advance. This program asks for an email address, validates it with the `in` operator, and takes everything before the `@` as a username:

````python
email = input("What's your email? ").strip().lower()

if "@" in email:
    username = email[:email.find("@")]
    print(f"Welcome, {username}!")
else:
    print(f"Sorry, {email} is not an email address.")
````

`email.find("@")` returns the index of the `@`, and a slice stops just before its `end`, so `email[:email.find("@")]` is every character before the `@`. If the user types `Reuben@MarcyLabSchool.org`, `lower()` turns it into `reuben@marcylabschool.org`, `find("@")` returns `6`, and `username` holds `"reuben"`.

When the input holds several pieces separated by the same characters, the `split` method breaks it into a list of pieces. Lists are the subject of chapter 1.7.

````python
print('apples, bananas, cherries'.split(', '))
# Output: ['apples', 'bananas', 'cherries']
````

Be careful when chaining after `split()`. Not every method returns a string: `split()` returns a list and `isdigit()` returns a boolean, so something like `'a, b'.split(', ').lower()` crashes with an `AttributeError`.

# 4. Output

The string tools we used on input work just as well on output, so a program can reshape a value before printing it: capitalize it, take a piece of it, or measure it. The more we can do with strings, the more control we have over exactly what our user reads.

One more useful tool for output is the `*` operator, which repeats a string, as you saw with `"3" * 5` in chapter 1.2. For example, this program prints a heading in capitals with a line of `=` beneath it:

````python
title = "Shopping List"

print(title.upper())
print("=" * len(title))
# Output:
# SHOPPING LIST
# =============
````

`title.upper()` returns a capitalized copy of the string and `"=" * len(title)` repeats `=` once for every character in the title. Since we are using the `len()` instead of hard coding the character amount the line stays exactly as long as the heading whenever `title` changes.

## Printing Several Values

Give `print()` several values separated by commas and it prints all of them on one line, with a space between each:

````python
name = "Ada"
age = 36

print("name:", name, "age:", age)
# Output: name: Ada age: 36
````

Two keyword arguments control the spacing. `sep` (separator) is what goes between the values, and its default is a space. `end` is what goes after the last value, and its default is `"\n"`, the new line character. That default is why every `print()` starts on a new line.

````python
print("a", "b", "c", sep="-")
# Output: a-b-c

print("loading", end="")
print("...", end="")
print("done")
# Output: loading...done
````

````python
print("1", "2", "3", sep="")
print(1, 2, 3, sep=" + ", end=" = 6\n")
````

Output:
```
123
1 + 2 + 3 = 6
```

`sep=""` puts nothing between the values so they print right next to each other. In the second print, `sep=" + "` goes between each number, and `end=" = 6\n"` replaces the default new line, so ` = 6` prints after the last value and then `\n` still moves us to the next line. `print()` took the numbers `1`, `2`, and `3` without needing `str()` because it turns every value into text before printing it.

## String Interpolation with f-strings

**f-strings** let us put values right inside a string. Put an `f` in front of the opening quote, and any expression inside curly braces `{}` gets **interpolated**, meaning Python works out its value and puts it into the string in that spot:

````python
print(f"The sum of 10 and 2 is {10 + 2}")
# Output: The sum of 10 and 2 is 12
````

Without the `f`, the curly braces are just regular characters and Python prints them exactly as they are:

````python
print("The sum of 10 and 2 is {10 + 2}")
# Output: The sum of 10 and 2 is {10 + 2}
````

Anything that produces a value can go inside the braces: a variable, arithmetic, a function call, a method call, an index, or a slice. That means every string tool from the Input section can shape a value right where it shows up in the message:

````python
first = "ada"
last = "lovelace"

print(f"Welcome, {first[0].upper()}{first[1:]}!")
# Output: Welcome, Ada!

print(f"Your initials are {first[0].upper()}.{last[0].upper()}.")
# Output: Your initials are A.L.

print(f"Your last name has {len(last)} letters.")
# Output: Your last name has 8 letters.
````

Each pair of braces holds its own expression. In the first message:

1. `first[0].upper()` takes the first character of the name (`"a"`) and capitalizes it (`"A"`)
2. `first[1:]` slices every character after the first (`"da"`)

Together they print the name with a capital letter even though `first` is stored in lowercase.

A slice is just as useful for leaving information out. A receipt should only show enough of a card number for the customer to recognize their card:

````python
card = "4111222233334444"

print(f"Paid with the card ending in {card[-4:]}")
# Output: Paid with the card ending in 4444
````

`card[-4:]` starts 4 characters from the end and goes to the end of the string, so it gives us the last 4 digits.

Like `print()` with commas, an f-string turns each value into text for us. Building the same message with `+` would need us to cast the number ourselves:

```py
"Your last name has " + str(len(last)) + " letters."
```

Forgetting the `str()` crashes the program with a `TypeError`, the same crash from Type Casting, because Python can't add a string and an integer. That's why f-strings are the way most Python programmers build a message: the words and values sit in the same order they'll be printed, and we never have to cast anything ourselves.

Predict, then run. Many receipts hide every digit of the card number except the last four.

````python
card = "4111222233334444"
print("*" * (len(card) - 4) + card[-4:])
````

Output: `************4444`

1. `len(card)` is `16`, so `len(card) - 4` is `12`
2. `"*" * 12` repeats `*` 12 times, one for every digit we're hiding
3. `card[-4:]` is `"4444"`, and `+` joins the two strings together. Both sides are already strings, so nothing needs casting.

The number of stars comes from the card number itself, so a 15-digit card would get 11 stars and still only show its last 4 digits.

## Formatting Numbers

Put a colon `:` after the expression inside the braces and you can say how the value should be displayed. This is called a **format specifier**. The one we'll use the most is `.2f`, which means "show this as a decimal number with exactly two digits after the decimal point":

````python
price = 5
print(f"${price:.2f}")
# Output: $5.00

total = 31.400000000000002
print(f"{total:.2f}")
# Output: 31.40
````

In `{price:.2f}`, everything before the colon is the value (`price`) and everything after it is how to display it (`.2f`). The first `$` is outside the braces, so it's just a regular character in the string.

`31.400000000000002` is the kind of number we get from arithmetic with decimals, because computers store decimal numbers in a way that is very slightly inexact.

The format specifier only changes how the number is **displayed**. The value stored in `total` is untouched. If we want to change the number itself, `round(total, 2)` returns `31.4`:

````python
total = 31.400000000000002

print(f"{total:.2f}")
# Output: 31.40  (displayed with 2 decimals)

print(total)
# Output: 31.400000000000002  (the value didn't change)

print(round(total, 2))
# Output: 31.4  (round() returns a new number)
````

# 6. Quick Reference

**Strings**

- Strings are **immutable**: methods never change the string, they return a new one. Store the result: `name = name.strip()`.
- Escape characters: `\"` is a quote inside a string, `\t` is a tab, `\n` is a new line.

**Input: clean → validate → parse**

| Step | What it means | Tools |
|---|---|---|
| Clean | Fix up what the user typed | `strip()`, `lower()`, `upper()`, `replace()`, `int()`, `float()` |
| Validate | Check it's acceptable before using it | `len()`, `in`, `startswith()`, `endswith()`, `isdigit()` |
| Parse | Take out the pieces we need | `string[index]`, `string[start:end]`, `find()`, `rfind()`, `split()` |

**String methods**

| Method | What it returns | Example |
|---|---|---|
| `s.strip()` | `s` without whitespace on either end | `'  hi '.strip()` → `'hi'` |
| `s.lower()` / `s.upper()` | `s` in all lowercase / all caps | `'YES'.lower()` → `'yes'` |
| `s.title()` | Every word capitalized | `'mary jane'.title()` → `'Mary Jane'` |
| `s.replace(old, new)` | Every `old` swapped for `new` (third argument limits how many) | `'555-1234'.replace('-', '')` → `'5551234'` |
| `s.startswith(x)` / `s.endswith(x)` | `True` if `s` begins / ends with `x` | `'https://a.org'.startswith('https://')` → `True` |
| `s.isdigit()` | `True` if every character is a digit | `'4.2'.isdigit()` → `False` |
| `s.find(x)` / `s.rfind(x)` | Index of the first / last `x`, or `-1` | `'bananas'.find('a')` → `1` |
| `s.split(sep)` | A list of the pieces between each `sep` | `'a, b'.split(', ')` → `['a', 'b']` |

**Lists: `.index()` instead of `.find()`**

| | Found | Not found |
|---|---|---|
| `string.find(x)` | The index | `-1` |
| `list.index(x)` | The index | `ValueError` crash, so check with `in` first |

**Indexing and slicing**

| Syntax | What it gives | `'Hello there!'` |
|---|---|---|
| `s[0]` | First character | `'H'` |
| `s[-1]` | Last character | `'!'` |
| `s[0:5]` | Index `0` up to but **not** including `5` | `'Hello'` |
| `s[:5]` / `s[6:]` | From the start / to the end | `'Hello'` / `'there!'` |
| `s[-6:]` | The last 6 characters | `'there!'` |
| `len(s)` | Number of characters, spaces included | `12` |

- An index only exists if `-len(s) <= index < len(s)`. Asking for any other index is an `IndexError`, including `s[0]` on an empty string.

**Type casting**

| Function | Example | Watch out |
|---|---|---|
| `int()` | `int("42")` → `42` | `int("hello")` and `int("4.2")` crash with `ValueError` |
| `float()` | `float("0.50")` → `0.5` | |
| `str()` | `str(1)` → `"1"` | |
| `bool()` | `bool("")` → `False` | Any non-empty string is `True`, even `"n"` |

- `input()` **always** returns a string. Validate with `isdigit()` first, then cast.

**Guard clauses for input**

```python
def next_year():
    age = input("How old are you? ").strip()

    if not age.isdigit():
        print(f"Sorry, {age} is not a number")
        return    # stop here so the bad input never reaches int()

    age = int(age)
    print(f"Next year you will be {age + 1}")
```

**Output**

| Syntax | What it does | Example |
|---|---|---|
| `print(a, b)` | Prints values with a space between | `print("age:", 36)` → `age: 36` |
| `sep=` | What goes **between** the values | `print(1, 2, sep="-")` → `1-2` |
| `end=` | What goes **after** the last value (default `"\n"`) | `print("hi", end="")` → no new line |
| `"=" * n` | Repeats a string `n` times | `"=" * len(title)` |
| `f"{x}"` | Puts the value of any expression into the string | `f"{10 + 2}"` → `12` |
| `f"{x:.2f}"` | Shows a number with 2 decimals (display only) | `f"{5:.2f}"` → `5.00` |

**Common traps**

| Trap | Fix |
|---|---|
| `answer.strip()` on its own line does nothing | `answer = answer.strip()` |
| `"YES" == "yes"` is `False` | Clean with `.lower()` before comparing |
| `int(input())` crashes on `twenty` | Store the string, check `isdigit()`, then cast |
| `x != "y" or x != "n"` is always `True` | `x != "y" and x != "n"` |
| `"1" + 1` crashes with `TypeError` | Use an f-string, or cast: `"1" + str(1)` |
| `'a, b'.split(', ').lower()` crashes | `split()` returns a list, not a string |
| `return print(...)` / `print` instead of `return` | Tests check what a function **returns**, so `print` gives back `None` |
