# 1. Key Terms

| **Terms** | **Definitions** |
|---|---|
| **DRY (Don't Repeat Yourself)** | Don't write the same code more than once. If the same lines appear in several places, write them once and reuse them. It's a principle of software engineering that aims to make code easier to maintain, update, and debug. |
| **Function** | A set of instructions that you give a name. Whenever you want those instructions to run, you "invoke" or "call" the function, which just means run it. Functions are named containers for statements that can be invoked. |
| **Built-in Functions** | Functions that Python has already defined and comes with, so they can be called without us defining them. `print()`, `len()`, and `type()` are all built-in functions. |
| **Docstring** | A string in triple quotes `"""` on the first line of a function that describes what the function does. We use it instead of a comment because a docstring stays attached to the function. You can view it by hovering over the function call, or with `help(function_name)`, which does not show comments. |
| **Parameters** | Placeholder names in the parentheses of a function's definition. Each one represents a value that we give to the function later on. They let the function do its job with any input, and change its behavior based on that input. |
| **Arguments** | The actual values we give when we call the function. |
| **Return Statement** | Used in a function to send a value back to the program so the program can use it. It also ends the function: no lines after `return` are executed. A function with no `return` returns `None`. |
| **Default Values** | A parameter can have a default value. If no argument is given for that parameter when the function is called, the default value is used instead. It is set in the definition: `def greet(name, greeting="Hello")`. |

| **TypeError** | Raised when a value is the wrong type for the operation, or when a function is called with the wrong number of arguments. |
| **Scope** | Defines where a variable can be accessed.<br>• **Global scope** is the file itself, outside of any function. Variables in the global scope can be used anywhere in the file.<br>• **Local scope** is the inside of a function. A variable or parameter created there exists only while the function runs and is reachable only within that function. |

# 2. What is `print()`?

`print()` is a **function**. It's already defined, and it outputs the values given to it in the Terminal so you can view what your program is doing. This is such a common task that the creators of programming languages provide a **built-in function** for it, so we don't have to create one ourselves.

# 3. Don't Repeat Yourself (DRY)

This code converts 100°C to Fahrenheit:

```python
boiling_point_C = 100
boiling_point_F = boiling_point_C * 9/5 + 32
print(boiling_point_F) # Output: 212.0
```

Suppose we also wanted to convert 0°C and 20°C to Fahrenheit in the same program. We could copy and paste the code and swap out the numbers like this:

```python
boiling_point_C = 100
boiling_point_F = boiling_point_C * 9/5 + 32
print(boiling_point_F) # Output: 212.0

freezing_point_C = 0
freezing_point_F = freezing_point_C * 9/5 + 32
print(freezing_point_F) # Output: 32.0

best_temperature_C = 20
best_temperature_F = best_temperature_C * 9/5 + 32
print(best_temperature_F) # Output: 68.0
```

This repeats the same code three times. It's messy and hard to read, and if the formula ever needed fixing, we would have to fix it in three places. This is where functions come into play.

# 4. Functions

A **function** is a reusable block of statements created to complete a task many times in a program. A function is used so we don't repeat code.

**Functions** are created with `def`, which is short for "define". It is followed by:
- A descriptive name
- **Parameters**, if any, inside parentheses `()`
- A colon `:`

````python
def convert_C_to_F(celsius):
    fahrenheit = celsius * 9/5 + 32
    print(fahrenheit)
````

The code is indented to show that it is *inside* the function block.

The three specific variable names from before (`boiling_point_C`, `freezing_point_C`, `best_temperature_C`) are replaced by one general parameter, `celsius`. The logic is exactly the same.

Running the cell above shows nothing. Defining a function only creates it. It doesn't run until it is called.

## Invoking a Function

A function can be **invoked**, or called, to execute its statements by typing the function name followed by its input values, or **arguments**, within parentheses `()`.

````python
convert_C_to_F(100)
convert_C_to_F(50)
convert_C_to_F(0)
````

## Docstrings

A **docstring** is a string on the first line of a function, written in triple double quotes `"""`, that gives a short description of what the function does. It is shown when you call `help()` on the function, and when you hover over the function's name in VS Code.

````python
def convert_C_to_F(celsius):
    """Print a Celsius temperature converted to Fahrenheit."""
    fahrenheit = celsius * 9/5 + 32
    print(fahrenheit)

help(convert_C_to_F)
````

## Function Calls Change the Control Flow

Code within a function only runs when the function is invoked, or called. When that happens, the interpreter jumps into the function and executes its first line.

````python
print('A')

def print_B():
    print('B')

print('C')

print_B()
print_B()
````

Here we can see that even though the function sits between `print('A')` and `print('C')`, `B` isn't printed until the function is invoked. We can tell where it is invoked by the function name followed by parentheses. If the function had parameters, we would have to include arguments in the `()`.

## Parameters and Arguments

````python
def say_hello():
    print("hello")

say_hello() # hello
say_hello() # hello
say_hello() # hello
````

The function `say_hello()` does the same thing every time because it has no **parameters**.

Parameters are set inside the `()` when defining the function. They are placeholders that reference the arguments, which are the values in the `()` when we invoke the function.

| | Parameter | Argument |

## Return Statements

The functions we have gone through use `print()` to show the result in the Terminal, but that result can't be used later in the program.
For a function to produce a value that can be reused by the program, you must use a **`return`** statement.

**`return`** does two things:

1. It returns a value to where the function was called.
2. It terminates the execution of the function, so no more lines in the function run after the `return` statement.

````python
def add(x, y):
    return x + y

total = add(5, 3) # The value of 5 + 3 is returned, resolving to `total = 8`

print(total) # We can now use the computed value outside of the function
````

Here the function is defined with the parameters `x` and `y`, and it returns `x + y` back to the program. The `total` line calls the function with the arguments `5` and `3`, which resolves to `total = 8`, since the function added the two arguments in the call. The `print()` statement then prints the variable's value.

> **Note:** the lesson names this variable `sum`. I use `total` because `sum` is already the name of a built-in Python function, and reusing the name would overwrite it.

A function call can also be used as an argument. Python runs the inner call first, and its returned value is what gets passed to the outer call:

````python
print(add(12, add(5, 3)))
````

1. `add(5, 3)` resolves to `8`.
2. `add(12, 8)` resolves to `20`.
3. `print(20)` prints `20`.

### `print()` is not `return`

````python
def say_hello(name):
    print(f"Hello, {name}!")

result = say_hello("Ada")
print(result)
````

Two lines are printed here because the function contains `print()` rather than `return`. The `result` line calls the function with an argument, and since the function has a `print()`, it prints the greeting. `print(result)` then prints `None`: a function always gives back a value when called, and if there is nothing to return, it returns `None`.

| | `print()` | `return` |
|---|---|---|
| **What it does** | Shows a value in the Terminal | Hands a value back to the program |
| **Who it's for** | The person looking at the screen | The rest of the code |
| **Can the value be used afterward?** | No | Yes |

> **Common mistake:** writing `print()` where you meant `return`, then wondering why the variable you saved the result in is `None`.

## Default Values and Keyword Arguments

A parameter can be given a **default value**. If no argument is given for it in a function call, the function uses the **default value**.

````python
def say_hello(hobby="play video games", name="Xavier"):
    print(f"Hi, my name is {name}. I like to {hobby}!")

say_hello()

say_hello('code', 'Eddie')
say_hello(name='Hua', hobby='play MapleStory')
````

- `say_hello()` gives no arguments, so both default values are used.
- `say_hello('code', 'Eddie')` passes the arguments **by position**: the first goes to `hobby` and the second goes to `name`.
- `say_hello(name='Hua', hobby='play MapleStory')` passes them **by keyword**. Each argument names the parameter it belongs to, which removes the need to put them in order.

Parameters with default values must come after the parameters without them.

## `*args` and `**kwargs`

> Normally a function takes a fixed number of arguments, set by how many parameters it has. `add(x, y)` takes exactly two, and giving it any more raises a `TypeError`.
>
> - A parameter with one star, `*args`, **collects extra positional arguments**, meaning the ones without keywords. The `*` gathers them and bundles them into a **tuple**.
> - A parameter with two stars, `**kwargs`, **collects extra keyword arguments**. The `**` gathers every `name=value` argument and bundles them into a **dictionary** of `key: value` pairs.
>
> The stars are the part that matters. The names `args` and `kwargs` are only a convention. This is how `print()` accepts any number of values. I don't need to write these yet, only recognize them.

````python
def describe(*args, **kwargs):
    print(args)
    print(kwargs)

describe("Ada", 36, city="Brooklyn", hobby="code")
````

# 5. Functions Create Local Variable Scope

**Scope** refers to where a variable has been assigned, which defines where it can be used or reached.

A variable created inside a function is in that function's **local scope**. It can only be used and reached inside that function, and nowhere else.

> **This cell errors on purpose.**

````python
def print_x():
    # x is reachable anywhere in this function
    x = 10

    print(x)

print_x()

print(x)  # NameError: name 'x' is not defined
````

Variables can be assigned at these levels of scope, listed from least reachable to most reachable:

- **Local variables**: variables and parameters assigned within a function. They only exist while that function is running.
- **Global variables**: variables assigned in the file, outside of any function. These are reachable anywhere in the file.
- **Built-in names**: names provided by Python itself, like `print`, `len`, and `type`. These are reachable everywhere. Avoid reassigning them.

Scopes can contain other scopes. For example, a file can contain a function, and functions can contain inner functions. Variables declared in the "outer" scope can be referenced anywhere within the "inner" scopes, but not the other way around.

A function can read a global variable:

````python
my_name = "Xavier"          # global: reachable anywhere in this file

def say_hi():
    print(f"Hi, my name is {my_name}")   # the function can see the global variable

say_hi()
````

If you reference a variable that is out of scope, you get a `NameError`, because the variable does not exist where you are trying to use it.

```python
def greet_friend(friend):
    message = f"Hi, {friend}!"
    print(message)

def say_goodbye(friend):
    message = f"Bye, {friend}!"
    print(message)
```

We can have two variables with the same name as long as they are in different scopes. This is the entire point of scope! Each function gets its own `message`, and neither can see or overwrite the other's. Without scope, every function in a program would have to invent a unique name for every one of its variables.

As a best practice, create variables in the lowest scope where they are needed: local rather than global.

## The `global` Keyword

Reading a global variable inside a function is fine. Changing one is where trouble starts.

> **This cell errors on purpose.**

````python
count = 0

def increment():
    count = count + 1

increment()
print(count)
````

The moment Python sees `count = ...` inside `increment`, it decides that `count` is a **local** variable of `increment`. Then it tries to work out `count + 1` using that local variable, which has no value yet. It never looks at the global `count`. The error is an `UnboundLocalError`.

You can force Python to use the global one by writing `global count` as the first line of the function, but **don't**. A function that reaches out and changes variables outside of itself is hard to predict by reading it.

Instead, **pass the value in as a parameter and `return` the new value**:

````python
count = 0

def increment(number):
    return number + 1

count = increment(count)
print(count)
````

1. `increment(count)` passes the current value, `0`, into the function, where it is called `number`.
2. The function returns `0 + 1`, which is `1`. It never touches the global `count`.
3. `count = increment(count)` stores the returned `1` back into `count`. The update happens outside the function, where anyone reading the code can see it.

# 6. Putting It Together: `player_damage`

This is my own example. It uses almost everything from this lesson in one function: parameters, a default value, a docstring, local variables, and `return`.

**What it does:** a player in a game gets hit. The function works out the player's health and armor after the hit.

**The rules of the game:**

- Armor blocks damage first. Only the damage that the armor can't block is taken off the player's health.
- Armor gets worn down by the damage it takes.
- Health, armor, and damage can never go below `0`.

**The parameters:**

| Parameter | What it holds | Default value |
|---|---|---|
| `health` | The player's health before the hit | None, so it must be given |
| `damage` | How strong the hit is | None, so it must be given |
| `armor` | The player's armor before the hit | `0`, so a player with no armor can leave it out |

## Version 1: Using `if` Statements

````python
def player_damage(health, damage, armor=0):
    """
    Work out a player's health and armor after taking a hit.

    Armor blocks damage first. Health, armor, and damage can never go below 0.
    Returns the new health and the new armor.
    """
    actual_damage = damage - armor
    if actual_damage < 0:
        actual_damage = 0

    new_armor = armor - damage
    if new_armor < 0:
        new_armor = 0

    new_health = health - actual_damage
    if new_health < 0:
        new_health = 0

    return new_health, new_armor


print(player_damage(100, 20))
print(player_damage(100, 20, 30))
````

### What each part does

| Code | What it does |
|---|---|
| `def player_damage(health, damage, armor=0):` | Defines the function with three parameters. `armor=0` gives `armor` a default value. |
| `""" ... """` | The docstring. It describes the function and shows up with `help(player_damage)` or when hovering over the name. |
| `actual_damage = damage - armor` | Works out how much damage gets past the armor. |
| `if actual_damage < 0:`<br>`actual_damage = 0` | If the armor is stronger than the hit, the subtraction gives a negative number. A hit can't do negative damage, so it is set to `0`. |
| `new_armor = armor - damage` | Wears the armor down by the strength of the hit. |
| `if new_armor < 0:`<br>`new_armor = 0` | Armor can't go below `0`. |
| `new_health = health - actual_damage` | Takes the damage that got through off the player's health. |
| `if new_health < 0:`<br>`new_health = 0` | Health can't go below `0`. |
| `return new_health, new_armor` | Sends both results back to the program and ends the function. |

An `if` statement runs the indented line under it only when its condition is `True`. For example, `if actual_damage < 0:` means "only do the next line if `actual_damage` is less than zero". Lesson 1.4 covers `if` statements in full.

### Tracing `player_damage(100, 20)`

No armor is given, so `armor` uses its default value, `0`.

| Line | Calculation | Value |
|---|---|---|
| `actual_damage = damage - armor` | `20 - 0` | `20` |
| `if actual_damage < 0:` | `20 < 0` is `False`, so the line under it is skipped | `20` |
| `new_armor = armor - damage` | `0 - 20` | `-20` |
| `if new_armor < 0:` | `-20 < 0` is `True`, so `new_armor` is set to `0` | `0` |
| `new_health = health - actual_damage` | `100 - 20` | `80` |
| `if new_health < 0:` | `80 < 0` is `False`, so the line under it is skipped | `80` |
| `return new_health, new_armor` | | `(80, 0)` |

The player had no armor, so the full `20` damage came off their health.

### Tracing `player_damage(100, 20, 30)`

This time `armor` is `30`.

| Line | Calculation | Value |
|---|---|---|
| `actual_damage = damage - armor` | `20 - 30` | `-10` |
| `if actual_damage < 0:` | `-10 < 0` is `True`, so `actual_damage` is set to `0` | `0` |
| `new_armor = armor - damage` | `30 - 20` | `10` |
| `if new_armor < 0:` | `10 < 0` is `False`, so the line under it is skipped | `10` |
| `new_health = health - actual_damage` | `100 - 0` | `100` |
| `if new_health < 0:` | `100 < 0` is `False`, so the line under it is skipped | `100` |
| `return new_health, new_armor` | | `(100, 10)` |

The armor blocked the whole hit, so health stayed at `100` and the armor dropped from `30` to `10`.

### Returning two values

`return new_health, new_armor` sends back two values at once. Python bundles them into a **tuple**, which is a fixed group of values written in parentheses. That is why the output looks like `(80, 0)`.

To use the two values separately, **unpack** them into two variables. The first value goes to the first name, and the second value goes to the second name:

````python
new_health, new_armor = player_damage(100, 20, 30)

print(f"Health: {new_health}")
print(f"Armor: {new_armor}")
````

## Version 2: Using `max()`

Version 1 repeats the same pattern three times: do a subtraction, then check whether the answer went below zero. That is a sign the code can be made DRY.

`max()` is a built-in function that returns the largest of the values it is given:

- `max(5, 0)` returns `5`
- `max(-10, 0)` returns `0`

So `max(some_number, 0)` means "give me this number, but never less than `0`". It does the same job as the `if` check, in one line.

````python
def player_damage(health, damage, armor=0):
    """
    Work out a player's health and armor after taking a hit.

    Armor blocks damage first. Health, armor, and damage can never go below 0.
    Returns the new health and the new armor.
    """
    actual_damage = max(damage - armor, 0)
    new_armor = max(armor - damage, 0)
    new_health = max(health - actual_damage, 0)

    return new_health, new_armor


print(player_damage(100, 20))
print(player_damage(100, 20, 30))
````

Both versions give the same results. Version 2 is shorter and easier to read because each value is worked out on a single line.

| | Version 1 | Version 2 |
|---|---|---|
| **How it stops values going below `0`** | An `if` statement after each calculation | `max(..., 0)` around each calculation |
| **Lines of code for the calculations** | 9 | 3 |
| **Result** | The same | The same |

## Using It Over Several Hits

The function doesn't change any variables outside of itself. It takes the current health and armor in as arguments and returns the new values, which is the pattern from the `global` section. To update the player, store the returned values back into the same variables:

````python
health = 100
armor = 30

health, armor = player_damage(health, 20, armor)
print(f"After hit 1: health {health}, armor {armor}")

health, armor = player_damage(health, 20, armor)
print(f"After hit 2: health {health}, armor {armor}")

health, armor = player_damage(health, 20, armor)
print(f"After hit 3: health {health}, armor {armor}")
````

The first hit is fully blocked. The second hit breaks the armor and `10` damage gets through. By the third hit there is no armor left, so all `20` damage comes off health.

## How This Example Uses the Lesson

| Idea from this lesson | Where it appears in `player_damage` |
|---|---|
| DRY | One function handles every hit. Version 2 also removes the repeated `if` checks. |
| Parameters and arguments | `health`, `damage`, and `armor` are parameters. `100`, `20`, and `30` are arguments. |
| Default value | `armor=0` |
| Docstring | The text in `"""` at the top of the function |
| `return` | Sends back the new health and armor so the program can keep using them |
| Local scope | `actual_damage`, `new_armor`, and `new_health` exist only inside the function. `print(actual_damage)` outside it raises a `NameError`. |
| No `global` | Values go in as arguments and come back out with `return` |

**Scope**

| Scope | Where it's created | Where it can be used |
|---|---|---|
| Local | Inside a function (including parameters) | Only inside that function |
| Global | In the file, outside any function | Anywhere in the file |
| Built-in | Provided by Python | Everywhere |

**Errors**

| Error | What it means | Common cause |
|---|---|---|
| `NameError` | Python can't find anything with that name | A misspelled name, a variable that is local to another function, or a name used before it is assigned |
| `TypeError` | The value is the wrong type, or the argument count is wrong | Adding a string to a number, or calling a function with too few or too many arguments |
| `UnboundLocalError` | A function tried to use a local variable before giving it a value | Reassigning a global variable inside a function |

**Worth remembering**

- Defining a function doesn't run it. It only runs when it is called.
- A function with no `return` gives back `None`.
- No lines in a function run after `return`.
- Parameters with default values go after the parameters without them.
- Positional arguments go before keyword arguments in a call.
- A name must be assigned before the line that uses it runs.
- Don't use `global`. Pass the value in as a parameter and `return` the new value.
- `max(number, 0)` stops a number from going below `0`.
