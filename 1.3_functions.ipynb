# 1. Key Terms

| **Terms** | **Definitions** |
|---|---|
| **DRY (Don't Repeat Yourself)** | Meaning don't write the same code mor than once. If the same line appears in several places, write them once and reuse them by referencing. Its the principle of a software engineer aiming to make it easier to maintain, update, and debug code. |
| **Function** | A function is a set of instruction that you label in the program. Whenever you want these instructions in a function to run you "invoke" or call it which just means run. They're named containers for statements that can be invoked. |
| **Built-in functions** | These are functions that python already defined and came with that can be called without us defining them. `print()`, `len()`, and `type()` are all built-in functions. |
| **docstring** | A docstring is a string made of triple quotes """ """ in the first line of a function to describe what the function does. We use this instead of comments as docstring stays inside the function. You can view the docstring covering over the function call or help("function name") will also show you the docstring where it wont show you comments. |
| **Parameters** | They're placeholders in the functions definitions parentheses that represents a value that we input to the function later on. It lets the function do its job with any input as well as change the behavior based on the input. |
| **Arguments** | This is the value we input when we call the function. |
| **Return Statement** | Return statements are used in functions to produce a value back to the program for the program to use as well as terminate the function as `return` statements end the function and do not continue executing lines in the function after `return`. |
| **Default Values** | parameters can have default values set which means if the argument is blank for this parameter when calling the function it will use the fault value instead. The default value can be set by `def greet(parameter1, parameter2="default value")` |
| **NameError** | A name error is produced when you are referencing or using a name in any line that isn't assigned to anything at the state that the line runs, the name can be misspelled, belongs to a different function or its defined further down in the program. |
| **TypeError** | This is produced when trying to use the incorrect types for the operation or when you haven't given the required amount of arguments to parameters for a function. |
| **Scope** | Scope defines where we can we can access the variable. <br>• The **Global Scope** is the file itself outside of any function. Variables in the global scope can be used anywhere. <br>• **Local Scope** is the inside of a function. That variable or parameter exists and is only reachable in within the function it was created. |

# 2. What is `print()`?

`print` is a **function**. It's already defined and outputs the values given to it in the terminal, so you can view what your program is doing. This is such a common task that the creators of languages provide `built-in functions` for us to output so we don't have to create one. 

# 3. Don't Repeat Yourself (DRY)

Suppose we wanted to also convert 0°C and 20°C to Fahrenheit in the same program. We could just copy and paste the code and swap out the numbers like this:
Copy

```python
boiling_point_C = 100
boiling_point_F = boiling_point_C * 9/5 + 32
print(boiling_point_F) # Output: 212

freezing_point_C = 0
freezing_point_F = freezing_point_C * 9/5 + 32
print(freezing_point_F) # Output: 32

best_temperature_C = 20
best_temperature_F = best_temperature_C * 9/5 + 32
print(best_temperature_F) # Output: 68
```

This is repeating code and its mostly messy and not clear to read. Which is where functions come to play.

# 3. Functions

A **function** is a reusable block of statements created to complete a task, or set of instructions, many times in the program. A function is used so we don't repeat code. 

**functions** are created with `def` which means define and its followed by:
- A descriptive name
- **parameters** if any inside of parentheses `()`.
- A `colon :`
