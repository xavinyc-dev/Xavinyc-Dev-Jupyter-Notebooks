# 1. Cell Basics

A notebook is made of two kinds of cells.

| Cell type | Used for | How to add one |
|---|---|---|
| **Markdown** | Notes, headings, tables | Hover between cells, click **+ Markdown** |
| **Code** | Python you can run | Hover between cells, click **+ Code** |

| Action | How |
|---|---|
| Run a code cell or render a Markdown cell | `Shift+Enter` |
| Edit a Markdown cell | Double-click it |
| Leave a cell (command mode) | `Esc` |
| Change the selected cell to Markdown | `Esc`, then `M` |
| Change the selected cell to code | `Esc`, then `Y` |
| Add a cell above / below | `Esc`, then `A` / `B` |
| Delete the selected cell | `Esc`, then `D` twice |

# 2. Headings

Start a line with `#` and a space. More `#` symbols make a smaller heading.

**You type:**

```markdown
# Biggest heading
## Second level
### Third level
```

**You get:**

# Biggest heading
## Second level
### Third level

**Use it for:** `#` for the notebook title and major sections, `##` for each concept inside a section. Headings also appear in the Outline panel as a clickable table of contents.

**Watch out:** the space after `#` is required. `#Title` will not work.

---
# 3. Divider Line

Three dashes on their own line draw a horizontal line.

**You type:**

```markdown
---
```

**You get:**

---

**Use it for:** separating major sections.

**Watch out:** leave a blank line above it. If `---` sits directly under a line of text, that text becomes a heading instead.

---
# 4. Bold and Italic

Wrap text in asterisks, with no spaces between the asterisks and the text.

**You type:**

```markdown
**bold text**
*italic text*
***bold and italic***
```

**You get:**

**bold text**

*italic text*

***bold and italic***

**Use it for:** bold for key terms and important words; italic for light emphasis.

---
# 5. Inline Code

Wrap text in single backticks. The backtick key is at the top-left of the keyboard, under `Esc`.

**You type:**

```markdown
Use `len()` to get the length of `my_str`.
```

**You get:**

Use `len()` to get the length of `my_str`.

**Use it for:** anything you would type: commands, variable names, method names, and values like `True`.

---
# 6. Code Blocks

Put three backticks on a line above and below the code. This **shows** code without running it. Add a language name after the opening backticks to color it.

**You type:**

````markdown
```python
my_str = 'Hello world'
print(len(my_str))
```
````

**You get:**

```python
my_str = 'Hello world'
print(len(my_str))
```

Common language names: `python` for Python, `bash` for terminal commands. With no language name, the block is plain text.

**Use it for:** terminal commands, or code you want to show but not run. For Python you want to run, use a **code cell** instead.

# 7. Bullet Lists

Start each line with a dash and a space. Indent with two spaces for a sub-bullet.

**You type:**

```markdown
- First point
- Second point
  - A sub-point
- Third point
```

**You get:**

- First point
- Second point
  - A sub-point
- Third point

**Use it for:** a group of related points where order doesn't matter, like a list of rules.

---
# 8. Numbered Lists

Start each line with a number, a period, and a space.

**You type:**

```markdown
1. First step
2. Second step
3. Third step
```

**You get:**

1. First step
2. Second step
3. Third step

**Use it for:** steps that happen in order, or a table of contents.

---
# 9. Callout Boxes

Start a line with `>` and a space.

**You type:**

```markdown
> **Tip:** Use the `Tab` key to autocomplete.
```

**You get:**

> **Tip:** Use the `Tab` key to autocomplete.

**Use it for:** tips, warnings, and corrections you want to stand out. Starting with a bold label like **Tip:**, **Warning:**, or **Note:** makes the purpose clear.

# 10. Tables

Use pipes `|` to separate columns. The second row of dashes is required: it marks the row above as the header.

**You type:**

```markdown
| Method | What it does |
|---|---|
| `upper()` | All uppercase |
| `lower()` | All lowercase |
```

**You get:**

| Method | What it does |
|---|---|
| `upper()` | All uppercase |
| `lower()` | All lowercase |

**Rules:**
- Every row needs the same number of pipes.
- The header row and the dashes row are both required.
- The columns do not need to line up neatly in what you type.
- For more columns, add more pipes: three columns means `|---|---|---|` on the dashes row.

**Use it for:** key terms with definitions, and quick-reference summaries.

---
# 11. Links

Put the visible text in square brackets and the address in parentheses, with no space between them.

**You type:**

```markdown
[Python documentation](https://docs.python.org)
```

**You get:**

[Python documentation](https://docs.python.org)

**Use it for:** linking to the lesson or documentation your notes came from.

---
# 12. Line Breaks and Paragraphs

Pressing Enter once does **not** start a new line in the result. Leave a **blank line** to start a new paragraph.

**You type:**

```markdown
These two lines
join into one.

This is a new paragraph.
```

**You get:**

These two lines
join into one.

This is a new paragraph.

**Watch out:** this is the most common surprise in Markdown. If your text runs together, add a blank line.

# 13. Quick Reference

| You type | You get |
|---|---|
| `# Title` | Biggest heading |
| `## Section` | Smaller heading |
| `---` | Horizontal divider line |
| `**text**` | **Bold** |
| `*text*` | *Italic* |
| `` `text` `` | `Inline code` |
| Three backticks above and below | Code block |
| `- item` | Bullet point |
| `1. item` | Numbered list item |
| `> text` | Callout box |
| Pipes between columns | Table |
| `[text](address)` | Link |
| A blank line | New paragraph |

## The pattern for a notes notebook

1. One Markdown cell at the top with a `# Title` and a contents list.
2. For each major section: a Markdown cell starting with `---` and a `# Section` heading.
3. For each concept: a Markdown cell with a `## Concept` heading and a short explanation, followed by a code cell with a runnable example.
4. A `# Quick Reference` table at the end.

## Not Markdown, but easy to confuse

- **`#` inside a code cell** is a Python comment, not a heading. The same symbol means different things depending on the cell type.
- **Square brackets like `[file_name]`** in command notes are placeholders meaning "put your own value here". You do not type the brackets.
