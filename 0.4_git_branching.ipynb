# 1. Key Terms

| Term | Meaning |
|---|---|
| **Main Branch** | The main branch of a repository. Whenever anyone visits a repository on GitHub or clones it down, this is what they will see. |
| **Feature Branch** | A copy of a repository at a point in time that lets developers work on a feature without impacting the rest of the project. |
| **Merge** | To combine the commit history of two or more branches into one. |
| **Pull Request** | A request for another developer to pull down your branch and review your code. If they approve the changes, they will merge your branch into the main branch. |
| **Fork** | A copy of a repository that is disconnected from the main repository. It typically includes the entire commit history of the main repository at the time the fork was created. |

# 2. Keep the Main Branch Stable

A **feature branch** is a copy of the main branch at a point in time. It lets developers work on a feature without impacting the main branch. Branches can be saved on GitHub and are only merged into the main branch when the feature is completed.

Branching ensures that the main branch is always a **"stable"** version. When somebody looks at your repo or copies it, they see and get the main branch.

```
main      A --- B --------------- E     (stable; E is the merge)
                 \               /
feature           C --- D ------        (work happens here)
```

Each letter is a commit. The feature branch starts from `B`, gets its own commits `C` and `D`, and is merged back into main when the feature is done.

# 3. Seeing, Creating, and Switching Branches

## git branch
Shows all branches in the local repository. The branch you are currently on is marked with `*`.

```bash
git branch
```

## git branch [branch_name]
Creates a new branch. It does **not** switch you to it.

```bash
git branch add-login-page
```

## git checkout [branch_name]
Switches to a branch.

```bash
git checkout add-login-page
```

## git checkout -b [branch_name]
Creates a new branch **and** switches to it in one line.

```bash
git checkout -b add-login-page
```

> **Note on `-b` vs `-B`:** my notes listed both. Use lowercase `-b`. It creates a new branch and stops with an error if that name already exists. Capital `-B` will **reset** an existing branch with that name, which can throw away its commits.

## Branch names
Use lowercase words separated by hyphens that describe the work, like `add-login-page` or `fix-typo-in-readme`. Names cannot contain spaces.

# 4. Working on a Branch and Pushing It to GitHub

Once you are on your branch, the normal Git workflow applies: make changes, add, commit, push.

```bash
git checkout -b add-login-page      # create the branch and switch to it
# ...edit files...
git add -A                          # stage the changes
git commit -m "Add login page"      # commit them to the branch
git push                            # upload the branch to GitHub
```

## Setting an "upstream" branch
The first time you push a new branch, GitHub doesn't have a branch with that name yet, so plain `git push` stops with a message about no upstream branch. Run this once:

```bash
git push -u origin add-login-page
```

- `origin` is the name of the remote repository (GitHub)
- `-u` links your local branch to the new remote branch (sets the "upstream")

After that, plain `git push` works for this branch.

# 5. Pull Requests and Merging

## Pull request (on GitHub)
After pushing your branch, open a **pull request** on GitHub. It asks another developer to review your code. If they approve the changes, your branch is merged into the main branch.

This is the usual way to merge when working with other people.

## git merge [branch_name]
Merges a branch **into the branch you are currently on**. To bring a feature branch into main locally, switch to main first:

```bash
git checkout main            # switch to the branch that will receive the changes
git merge add-login-page     # merge the feature branch into it
```

## After a pull request is merged on GitHub
Your local main is now behind the remote. Update it:

```bash
git checkout main
git pull
```

---
# 6. Deleting a Branch

Once a branch has been merged, you no longer need it.

```bash
git branch -d add-login-page     # delete; refuses if the branch has unmerged work
git branch -D add-login-page     # force delete, even if the work was never merged
```

> **Warning:** capital `-D` deletes the branch even if its commits were never merged, so that work can be lost. Use lowercase `-d` unless you are sure.

You cannot delete the branch you are currently on. Switch to another branch first, such as `main`.

# 7. Quick Reference

| Command | What it does |
|---|---|
| `git branch` | See all branches in the local repository |
| `git branch [branch_name]` | Create a new branch |
| `git checkout [branch_name]` | Switch to a branch |
| `git checkout -b [branch_name]` | Create a new branch and switch to it |
| `git push -u origin [branch_name]` | Push a new branch to GitHub for the first time |
| `git merge [branch_name]` | Merge a branch into the current branch |
| `git branch -d [branch_name]` | Delete a branch (safe) |
| `git branch -D [branch_name]` | Force delete a branch |

## The full feature branch cycle

```bash
git checkout main                    # start from main
git pull                             # make sure main is up to date
git checkout -b my-feature           # create and switch to a feature branch
# ...edit files...
git add -A
git commit -m "Describe the work"
git push -u origin my-feature        # push the branch to GitHub
# open a pull request on GitHub, get it reviewed and merged
git checkout main
git pull                             # bring the merged work into local main
git branch -d my-feature             # delete the finished branch
```
