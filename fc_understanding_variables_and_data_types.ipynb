# 1. Variables

Variables are labeled boxes used for storing and referencing data of different types, such as numbers or text.

To create a variable, write its name on the left, then the assignment operator (`=`), then the value you want to store on the right.

```python
name = 'John Doe'
age = 25

print(name)
print(age)
```

Here `name` is the variable that holds the value `'John Doe'`.

That value is a **string**. Strings are a series of characters used to represent text, written in single or double quotes.

---
# 2. Naming Rules and Conventions

## Rules
Breaking any of these raises a `SyntaxError`.

- Names can **only start** with a letter or an underscore (`_`), **not** a number
- Names can only contain alphanumeric characters (`a-z`, `A-Z`, `0-9`) and underscores (`_`)
- Names can **not** be one of Python's reserved keywords, such as `if`, `class`, and `def`

Names are also **case-sensitive**: `age`, `Age`, and `AGE` are three different variables.

```python
age = 25
Age = 30
AGE = 35

print(age, Age, AGE)   # three separate variables
```

```python
_valid_name = 'starts with an underscore'
valid_name_2 = 'numbers are fine after the first character'
print(_valid_name)
print(valid_name_2)

# Uncomment any line below to see the SyntaxError:
# 2nd_name = 'starts with a number'
# my-name = 'contains a hyphen'
# class = 'reserved keyword'
```

## Snake case
Variable names should be lowercase, with words separated by underscores. This is called **snake case**.

```python
my_variable_name = 'freeCodeCamp'
print(my_variable_name)
```

## Be descriptive
Choose names that say what the value is. To store a user's age, `user_age` is better than `age` or an abbreviation.

Avoid single-letter names such as `x`, because they communicate no purpose or meaning.

```python
x = 25           # unclear: 25 what?
user_age = 25    # clear

print(user_age)
```

# 3. Comments

A comment starts with the pound symbol (`#`). Python ignores everything after the `#` on that line.

```python
# This is a single-line comment
print('Comments are ignored')  # a comment can also follow code on the same line
```

# 4. Common Data Types

## Integer
A whole number without decimals, for example `10` or `-5`.

```python
my_integer_var = 10
print('Integer:', my_integer_var)
```

## Float
A number with a decimal point, like `4.41` or `-0.4`.

```python
my_float_var = 4.50
print('Float:', my_float_var)
```

## String
A sequence of characters enclosed in single or double quotation marks, like `'Hello world!'`.

```python
my_string_var = 'hello'
print('String:', my_string_var)
```

## Boolean
A true or false type, written as `True` or `False` (capitalized).

```python
my_boolean_var = True
print('Boolean:', my_boolean_var)
```

---
# 5. The `print()` Function

`print()` is a **built-in function**, which means you can use it without defining it yourself. It sends text and other values to the terminal.

The value inside the parentheses is called an **argument**. Here the string `'Hello world!'` is the argument passed to `print()`.

```python
print('Hello world!')
```

## Multiple arguments
`print()` can show several values separated by commas. Python automatically puts a space between them.

```python
name = 'John Doe'
age = 25

print('Name:', name, 'Age:', age)
```

# 6. `type()` and `isinstance()`

## type()
Shows the class, or type, of a variable.

```python
developer = 'Devin'
print(type(developer))
```

`<class 'str'>` means `developer` is a string.

Here are all four data types so far:

```python
print(type(my_integer_var))
print(type(my_float_var))
print(type(my_string_var))
print(type(my_boolean_var))
```

## isinstance()
Checks whether a variable matches a specific data type. It takes a value and the type to check against, then returns a Boolean.

This is useful when you need to verify a variable's type before performing operations on it.

```python
print(isinstance(my_integer_var, int))
print(isinstance(my_integer_var, str))
print(isinstance(my_string_var, str))
```

## Checking multiple types at once
Pass the types inside parentheses (a tuple). It returns `True` if the value matches **any** of them.

```python
print(isinstance(my_float_var, (int, float)))
print(isinstance(my_string_var, (int, float)))
```
